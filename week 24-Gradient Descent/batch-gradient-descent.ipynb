### intercept 

slope of the intercept = $\frac{\partial L}{\partial B_0} = \frac{-2}{n} \sum_{i=1}^{n}(y_i - \hat{y_i})$           
Question : what is $y_i$ ?          
$y_i$ is basiclllay output column . meaning target variable  = hamara training data ma jo output column ha y_train ha           
Question : what is $\hat{y_i}$ ?            
$\hat{y_i}$ is apka model jo input data ka upar predication karega. example jab aap 353 rows deta ho tho apko
tho vo ek predication karega 353 predication kareaga        
deko tho             
$y_i$ is y_train = matrix of 353 rows and 10 columns        
$\hat{y_i}$ is y_test = matrix of 353 rows and 1 column         
Question : What is the meaning of $\frac{1}{n} \sum_{i=1}^{n} X{i}$ ya par $X_i = (y_i - \hat{y_i})$            
matlab aap iska mean nikal raha ho

observe the Sequence for slope calcuation 
- $(y_i - \hat{y_i})$ ek matrix ayaga with 353 rows
- ab hum us matrix ma mean nikalaga ga ek scaler ayega
- us scaler ko ab hum -2 sa multiply karega 
- fir us scaler ko learning rate ek scaler sa multiply kareag
- fir old intercept value sa us scaler ko subtract kareaga
- meaning our intercept parameter is always a scaler value

```python
class GradientDescantRegressor:

    def __init__(self,learning_rate=0.001,epoches = 100):
        self.intercept_ = None
        self.coef_ = None
        self.learning_rate = learning_rate
        self.epoches = epoches
        
    def fit(self,X_train,y_train):
        # intialise the coeffs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        # lakin muja kesa pata hoga ki kitna coefficients ha 
        # tho X_train ka shape bataya ga number of coefficients
        self.coef_ = np.ones(X_train.shape[1])
        for i in range(self.epoches):
            # update all the coeff and the intercept
            y_hat = np.dot(X_train,self.coef_) + self.intercept_
            slope_of_intercept  = -2 * np.mean(y_train - y_hat)
            self.intercept_ = self.intercept_- (self.learning_rate * slope_of_intercept) 
        
        print('intercept :',self.intercept_)
        print('coefficents :',self.coef_)
        
    def predict(self,X_test):
        pass
# lets check by creating our object
gd_object = GradientDescantRegressor()
gd_object.fit(X_train,y_train)
```

### coefficents

slope of the coefficent = $\frac{\partial L}{\partial B_m} = \frac{-2}{n} \sum_{i=1}^{n}(y_i - \hat{y_i})X_{mn}$

```python
class GradientDescantRegressor:

    def __init__(self,learning_rate=0.001,epoches = 100):
        self.intercept_ = None
        self.coef_ = None
        self.learning_rate = learning_rate
        self.epoches = epoches
        
    def fit(self,X_train,y_train):
        # intialise the coeffs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epoches):
            # update all the coeff and the intercept
            y_hat = np.dot(X_train,self.coef_) + self.intercept_
            slope_of_intercept  = -2 * np.mean(y_train - y_hat)
            self.intercept_ = self.intercept_- (self.learning_rate * slope_of_intercept) 

            slope_of_coefficents  =  -2 * np.dot((y_train - y_hat),X_train)/X_train.shape[0]
            self.coef_ = self.coef_ - (self.learning_rate * slope_of_coefficents )
        print('intercept :',self.intercept_)
        print('coefficents :',self.coef_)
        
    def predict(self,X_test):
        pass

gd_object = GradientDescantRegressor()
gd_object.fit(X_train,y_train)
```

```python
class GradientDescantRegressor:

    def __init__(self,learning_rate=0.001,epoches = 100):
        self.intercept_ = None
        self.coef_ = None
        self.learning_rate = learning_rate
        self.epoches = epoches
        
    def fit(self,X_train,y_train):
        # intialise the coeffs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epoches):
            # update all the coeff and the intercept
            y_hat = np.dot(X_train,self.coef_) + self.intercept_
            slope_of_intercept  = -2 * np.mean(y_train - y_hat)
            self.intercept_ = self.intercept_- (self.learning_rate * slope_of_intercept) 

            slope_of_coefficents  =  -2 * np.dot((y_train - y_hat),X_train)/X_train.shape[0]
            self.coef_ = self.coef_ - (self.learning_rate * slope_of_coefficents )
        print('intercept :',self.intercept_)
        print('coefficents :',self.coef_)
        
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_

```

```python
gd_object = GradientDescantRegressor(epoches=1000,learning_rate=0.5)
gd_object.fit(X_train,y_train)
```

```python
y_pred = gd_object.predict(X_test)
```
