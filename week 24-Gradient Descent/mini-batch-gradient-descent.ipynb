# Mini-batch class

We use our old batch gradient descant class

```python
class GradientDescantRegressor:

    def __init__(self,learning_rate=0.001,epoches = 100):
        self.intercept_ = None
        self.coef_ = None
        self.learning_rate = learning_rate
        self.epoches = epoches
        
    def fit(self,X_train,y_train):
        # intialise the coeffs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epoches):
            # update all the coeff and the intercept
            y_hat = np.dot(X_train,self.coef_) + self.intercept_
            slope_of_intercept  = -2 * np.mean(y_train - y_hat)
            self.intercept_ = self.intercept_- (self.learning_rate * slope_of_intercept) 

            slope_of_coefficents  =  -2 * np.dot((y_train - y_hat),X_train)/X_train.shape[0]
            self.coef_ = self.coef_ - (self.learning_rate * slope_of_coefficents )
        print('intercept :',self.intercept_)
        print('coefficents :',self.coef_)
        
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_

```

```python
import random

class MiniBatchGradientDescantRegressor:

    def __init__(self,batch_size,learning_rate=0.001,epoches = 100):
        self.intercept_ = None
        self.coef_ = None
        self.learning_rate = learning_rate
        self.epoches = epoches
        self.batch_size = batch_size
        
    def fit(self,X_train,y_train):
        # intialise the coeffs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epoches):
            for j in range(int(X_train.shape[0]/self.batch_size)):
            # update all the coeff and the intercept
                random_rows_index = random.sample(range(X_train.shape[0]),self.batch_size)
                y_hat = np.dot(X_train[random_rows_index],self.coef_) + self.intercept_
                slope_of_intercept  = -2 * np.mean(y_train[random_rows_index] - y_hat)
                self.intercept_ = self.intercept_- (self.learning_rate * slope_of_intercept) 

                slope_of_coefficents  =  -2 * np.dot((y_train[random_rows_index] - y_hat),X_train[random_rows_index])/X_train.shape[0]
                self.coef_ = self.coef_ - (self.learning_rate * slope_of_coefficents )
        print('intercept :',self.intercept_)
        print('coefficents :',self.coef_)
        
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_

```

```python
import random

class MBGDRegressor:
    
    def __init__(self,batch_size,learning_rate=0.01,epochs=100):
        
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        
    def fit(self,X_train,y_train):
        # init your coefs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            
            for j in range(int(X_train.shape[0]/self.batch_size)):
                
                idx = random.sample(range(X_train.shape[0]),self.batch_size)
                
                y_hat = np.dot(X_train[idx],self.coef_) + self.intercept_
                #print("Shape of y_hat",y_hat.shape)
                intercept_der = -2 * np.mean(y_train[idx] - y_hat)
                self.intercept_ = self.intercept_ - (self.lr * intercept_der)

                coef_der = -2 * np.dot((y_train[idx] - y_hat),X_train[idx])
                self.coef_ = self.coef_ - (self.lr * coef_der)
        
        print(self.intercept_,self.coef_)
    
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
mbr = MBGDRegressor(batch_size=int(X_train.shape[0]/50),learning_rate=0.01,epochs=100)
mbr.fit(X_train,y_train)
```

```python
y_pred = mbr.predict(X_test)
print('r2_score :',r2_score(y_test,y_pred))
```
