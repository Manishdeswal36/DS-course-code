### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
first_term = int(input('enter the num'))
second_term = int(input('enter the num'))
d = second_term - first_term # commen differnece
# number of terms 
nth_term = int(input('enter the number of term you want'))

# a1 = 1
# a2 = 3
# d = a2-a1 = 2
# an = a +(n-1)d
# or simplify an = nd + a-d
# 1,3,5,7... odd number nth term is 
# write equation for nth term 
an = first_term + (nth_term-1)*d
print('the {}th term is '.format(nth_term),an)
````
