# `is` Keyword
In Python, the `is` keyword is used for identity testing and object comparison. It checks if two objects refer to the same memory location.



```python
a = [1, 2, 3]
b = a  # Both a and b refer to the same list object
c = a.copy()
print('id of a ',id(a))
print('id of b ',id(b))
print('id of c ',id(c))
print('----------------')
print('equal operator between a==b:', a==b)
print('equal operator between a==c:', a==c)
print('----------------')
print('id of a & b :',a is b)  # True, because a and b refer to the same object
print('id of a & c :',a is c)  # values are same but object address is different 
```

## isalnum()
Returns True if all characters in the string are alphanumeric, otherwise False.

```python
"abc123".isalnum()  # True
"abc 123".isalnum()  # False (contains a space)
```

## isalpha()           
Returns True if all characters in the string are alphabets, otherwise False.

```python
"abc".isalpha()  # True
"abc123".isalpha()  # False (contains a number)
```

## isascii

Returns True if all characters in the string are ASCII, otherwise False.


```python
"hello".isascii()  # True
"你好".isascii()  # False (contains non-ASCII characters)
```

## isdecimal()

Returns True if all characters in the string are decimal (consisting of decimal digits), otherwise False.

```python
"12345".isdecimal()  # True
"12.345".isdecimal()  # False (contains a dot)
```

## isidentifier()

Returns True if the string is a valid Python identifier, otherwise False.

```python
"variable_name".isidentifier()  # True
"123abc".isidentifier()  # False (starts with a number)
```

## islower()

Returns True if all alphabetic characters in the string are lowercase, otherwise False.

```python
"hello".islower()  # True
"Hello".islower()  # False (contains uppercase)
```

## isupper()

Returns True if all alphabetic characters in the string are uppercase, otherwise False.

```python
"HELLO".isupper()  # True
"Hello".isupper()  # False (contains lowercase)
```

.isspace()

Returns True if all characters in the string are whitespace, otherwise False.

```python
"   ".isspace()  # True
"   hello   ".isspace()  # False (contains non-whitespace characters)
```

## isprintable()

Returns True if all characters in the string are printable, otherwise False

```python
"hello".isprintable()  # True
"\n".isprintable()  # False (contains a non-printable character)
```

## istitle()

Returns True if the string is a titlecased string (words start with uppercase and the rest are lowercase), otherwise False.

```python
"This Is A Title".istitle()  # True
"This is not a Title".istitle()  # False
```
