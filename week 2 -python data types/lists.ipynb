# List comprehension

List Comprehension provides a concise way of creating lists.
Advantages of List Comprehension
- More time-efficient and space-efficient than loops.
- Require fewer lines of code.
- Transforms iterative statement into a formula.

## by for loop 

````python
# list = []
# for (iter) in (iterable):  # test
#    list.append(output expression)
number = []
for iter_var in range(1,11):
    number.append(iter_var) # you do any operation here 
number
````

````python
number = []
for iter_var in range(1,11):
    number.append(iter_var**2) # square         # you do any operation here 
number
````

## list com to create list 

````python
# [(output expression) for (iter) in (iterable)]

#   (operation in iter) 

[iter_var**2 for iter_var in range(1,11)]
# same 
list_1 = [numbers**2 for numbers in range(1,11)]
print(list_1)
````

## Nested if with List Comprehension
basket = ['apple','guava','cherry','banana']
my_fruits = ['apple','kiwi','grapes','banana']

 add new list from my_fruits and items if the fruit exists in basket and also starts with 'a'

``` [[expression for inner_element in inner_list] for outer_element in outer_list] ```


````python
basket = ['apple','guava','cherry','banana']
my_fruits = ['apple','kiwi','grapes','banana']
# dona list ma sa vo btana ho jo a sa start ho raha 
# check for basket 
[ i for i in basket if i.startswith('a')]
#check in my_fruit
[j for j in my_fruits if j.startswith('a')]

# dono ma alga iter_var chan na acha ha ek karo dona ma 
fruit = [i for i in basket for i in my_fruits if i.startswith('a')]
print(fruit)
````

````python
[iter_var for iter_var in my_fruits if iter_var in basket if iter_var.startswith('a')]
#        1.  loop in my_fruit lits    2.# condition in basket     3.  # ask the conditon 
#   yaha hua kya -->
#  1. myfruit ma loop  chalya 
#  2. aab humana pucha ki vo element jo my_fruit or basket  list ma deko zara in means
#  belongs to ha tho 
# #  3. fir un par conditon     
# final our anser                 
````

# Zip

The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

````python
# item by item 2 iterables ko ek sath pack kar deta ha 
# you can convert any data_type(zip object)
# i do tuple because in list \
# pyalce show er
````

## add two list by item wise

````python
l1 = [1,2,3,4]
l2 = [4,5,6,7]
zip(l1,l2)
list(zip(l1,l2)) # type: ignore
````

````python
print(list(zip(l1,l2)))# type: ignore
````

````python
[i+j for i,j in list(zip(l1,l2))] # type: ignore
````
