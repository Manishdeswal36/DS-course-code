###`Q-2:` Write a Pandas program to add, subtract, multiple and divide two Pandas Series.

````python
# so muja function bana input ma two series la aor putput ma arithmetic 
# operation perform kara 

def perform_arithmatic(series_1:pd.Series,series_2:pd.Series)-> pd.Series:
    print('--Do addition--')
    print(series_1 + series_2)
    print("-- Do subtraction-- ")
    print(series_1 - series_2)
    print("-- Do mul-- ")
    print(series_1 * series_2)
    print("-- Do division-- ")
    print(series_1 / series_2)
    
a = pd.Series([1,2,3])
b = pd.Series([4,5,6])
perform_arithmatic(a,b)
````

````python
# so to convert dollar to rupees
# 1 dollar = 80  rupees simple
items_in_rupee = new_items* 80
items_in_rupee
````
