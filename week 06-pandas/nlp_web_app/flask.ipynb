# hello world program 

```python
from flask import Flask

app = Flask(__name__)
# this is our decorator add funcality connect with the router
# isko  ek function dega jiska name index hi hota ha geneally 
@app.route('/') 
def index():
    return "Hello World"
# a run function to run the python file 
app.run()
```

```python
# class ka andr sabsa jada import ha ya route inko app url samaj sakta ho 
# jasa hi apki Website ma koi slash mara ga 
# routue yani url ko hit karo aor ya string bhej do 
```

## heading tag in html

```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def index():
    return "<h1>Hello World</h1>"
# tho koi bhi text ko h1 tag ka andr dalo
# vo heading ban jata ha 
app.run(debug = True)
```

```python
# ya jo mera pura app.py file ha ya server par ha 
# aor iska bhi reponse ha 
# ya utad kar ka client ka computer par ja raha ha 
# jo ki hamara browser ha 
# tho aap index func ma return ka value change karka 
# ala alag values ko bhaj sakta hao 
```
