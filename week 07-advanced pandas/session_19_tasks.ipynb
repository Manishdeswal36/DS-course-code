### `Q-4:` Perform `groupby` based on the `"Team"` column and then perform Z Normalization on top of the below columns of each group:
1. Passes
2. Passes Completed
3. Attempted Line Breaks
4. Completed Line Breaks

You have to make a python function named `z_normalization` which takes two arguments:

1. *group:* Every group that you have created
2. *cols_to_perform:* This parameter takes a list of columns on which you have to perform the Z-Normalization.

For this problem, you have to use th `apply()` method.

$\Large Z \ - \ Normalization = \frac{X_i - \mu}{std} $

After that find out the below values for each group:
- minimum "Passess"
- maximum "Passess"
- minimum "Yellow Cards"
- maximum "Yellow cards"
- average "Yellow Cards"
- maximum "Attempted Line Breaks"
- minimum "Attempted Line Breaks"
- standardard deviation of "Attempted Line Breaks"
- average Possession

````python
import pandas as pd

fifa_df = pd.read_csv("https://docs.google.com/spreadsheets/d/e/2PACX-1vT3D_x_4DS6d51LKJ7ze1sxT5WpV5uiSVOFYHLwBiGru6vFyVv5h5-83AwFjxWYiWfCDjDAaarHAV-k/pub?gid=0&single=true&output=csv")
fifa_df
````

````python
def z_normalisation(group_col:pd.DataFrame,coloumn:pd.Series)->pd.DataFrame:
    group = fifa_df.groupby(group_col)
    
    group['z_normalised score'] = ( group - group[coloumn].mean())/group[coloumn].std()
    return group.apply()
z_normalisation(group_col='Team',coloumn='Passes')
````

````python
def z_normal(group):
    group['z_normal '] = ( group - group[coloumn].mean())/group[coloumn].std()
    return group
````

````python
cols_to_perform = ['Passes', 'Passes Completed', 'Attempted Line Breaks','Completed Line Breaks']
# this are my coloums to perform z-normlisation 
# perform groupby on team
fifa_df[cols_to_perform]
groups = fifa_df.groupby('Team')
# now make our z_normlisation function 
def z_normalisation(group:pd.DataFrame,coloumns_to_perform:pd.Series) -> pd.DataFrame:
    for coloumn in coloumns_to_perform:
        std = group[coloumn].std()
        mean = group[coloumn].mean()
        group[f'{coloumn}_z_norm'] = (group[coloumn] - mean)/std
        return group

groups.apply(z_normalisation,coloumns_to_perform=cols_to_perform)
````

````python
# groupby karo aor multiindex use karka multiple  aggreate option lagao
fifa_df.groupby('Team').agg({
    'Passes':['min','max'],
    'Yellow Cards':['min','max','mean'],
    'Attempted Line Breaks':['min','max','std']
})
    
````

###`Q-6` Most Successful bowler against any batsman. Find that pair of bowler and batsman.
> Most Successful in terms of dissmissal. A bowler who have dissmissed any batsman most no of times. If any two pairs have same no of dissmisal, consider runs conceded by bowler to that batsman. Those who have concede lesser runs is more successful.

in cricket, "dismissed" means that the batsman has lost their wicket.
"A bowler has dismissed a batsman X times," it means that the bowler has taken the wicket of that batsman X times in different matches or innings.

````python
ipl.columns
````

````python
# so muja nikalan ha batsman bowler pair jisma bowler na wicket max li ho and runner minimium ho 
pair_df = ipl.groupby(by=['batter','bowler']).agg({'isWicketDelivery': 'count','batsman_run': 'sum'}).reset_index()
# ab hum chata ha wicket max (descending) sort ho and batsman_run min(ascending) ho 
pair_df.sort_values(by=['isWicketDelivery','batsman_run'],ascending=[False,True])
````
