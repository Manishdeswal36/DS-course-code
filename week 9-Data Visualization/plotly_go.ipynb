Figure Data Structure
At the heart of Plotly is the Figure object, which represents a complete visualization.

A Figure consists of three main parts:

- data: A list of "traces" (graphical objects like scatter plots, bar charts, etc.).

- layout: Configuration of the figure’s appearance (titles, axes, colors, annotations).

- frames: For animations, sequences of frames defining changes over time.

Figures are internally represented as nested dictionaries or JSON-like structures, which are passed to Plotly.js (the underlying JavaScript engine) for rendering

# Scatter plot 

```python
# Problem : we are going to draw a scatter plot between 
# Batsman Avg( x axis) and Batsman Strike rate(Y axis )
# of top 50 batsman in ipl(all time )
```

```python

top50 = ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(50).index.tolist()
new_ipl = ipl[ipl['batsman'].isin(top50)]

new_ipl
```

```python
# Calculating SR
# SR = (number of runs scored )/(number of balls played) * 100

runs = new_ipl.groupby('batsman')['batsman_runs'].sum()
balls = new_ipl.groupby('batsman')['batsman_runs'].count()

strike_rate = (runs/balls)*100
strike_rate = strike_rate.reset_index()
strike_rate
# ya col ka name batsman runs nahi strike rate hona chya tha 
```

```python
strike_rate.rename( columns= {'batsman_runs': 'strike_rate'},inplace=True)
```

```python
strike_rate
```

```python
# Calculating Avg
# Avg = (Total numbers of runs )/ (number of outs)

# Calculating number of outs for top 50 batsman

out = ipl[ipl['player_dismissed'].isin(top50)]

nouts = out['player_dismissed'].value_counts()

avg = runs/nouts

avg = avg.reset_index()
avg
```

```python
avg.rename(columns={'index':'batsman',0: 'avg'},inplace = True)
```

```python
avg
```

```python
# merge the boths 

batsman_df = avg.merge(strike_rate,on='batsman')
batsman_df
```

```python
import plotly.graph_objects as go

# Convert to string if not already
batsman_df['batsman'] = batsman_df['batsman'].astype(str)
batsman_df['batting_team'] = batsman_df['batting_team'].astype(str)

# Calculate sizeref for bubble size scaling
sizeref = 2. * max(batsman_df['six']) / (40. ** 2)

trace = go.Scatter(
    x=batsman_df['avg'],
    y=batsman_df['strike_rate'],
    mode='markers',
    marker=dict(
        size=batsman_df['six'],
        sizemode='area',
        sizeref=sizeref,
        color=batsman_df['batting_team'].astype('category').cat.codes,  # Map categories to numbers
        colorscale='Viridis',
        showscale=True,
        line=dict(width=1, color='DarkSlateGrey')
    ),
    customdata=batsman_df[['batsman', 'batting_team']],  # Pass extra info for hover
    hovertemplate=(
        "<b>Batsman:</b> %{customdata[0]}<br>" +
        "<b>Team:</b> %{customdata[1]}<br>" +
        "<b>Avg:</b> %{x}<br>" +
        "<b>Strike Rate:</b> %{y}<br>" +
        "<extra></extra>"  # Hide trace info box
    )
)

layout = go.Layout(
    title='Bubble Chart with Hover Info',
    xaxis={'title': 'Average'},
    yaxis={'title': 'Strike Rate'}
)

fig = go.Figure(data=[trace], layout=layout)
fig.show()
```
