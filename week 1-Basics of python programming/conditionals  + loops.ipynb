Guessing game

````python
#  random number generate hua 
#  user sa number liya fir us number ko match kiya ki apka guess kitna jada ha kitna kam ha
# by conidtionals
````

````python
import random

while True:
    rand_no = random.randrange(1,10)
    print("the random no is ",rand_no)

    user_no = int(input("enter your number"))
    print("your number is ", user_no)

    if user_no > rand_no:
        high = user_no - rand_no
        print("your guess is high by",high)

    if user_no < rand_no:
        low = rand_no - user_no
        print("your guess is high by ",low)
    print('----------------------')
    if user_no == rand_no:
        break
print( "you win")
````

using nth term formula 

````python
# using nth term forumula
def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        fib_series = [0, 1]
        [fib_series.append(fib_series[-1] + fib_series[-2]) for _ in range(2, n)]
        return fib_series

# Number of terms to generate in the Fibonacci series
n_terms = 10

# Generate and print the Fibonacci series using list comprehension
fib_series = fibonacci(n_terms)
print(fib_series)
# last term + last second term ko add kiya aor append kar diya empty list ma
````

nth term + list compreshension +nested loop 

````python
# Define the number of terms in the Fibonacci series
n = 10

# Create a list comprehension to generate the Fibonacci series
fib_series = []
fib_series = [0 if i == 0 else 1 if i == 1 else (fib_series[i-1] + fib_series[i-2]) for i in range(2,n_terms)]

# Print the Fibonacci series
print(fib_series)
````
