## Import libraties 

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

## Import dataset

```python
data = pd.read_csv('heart.csv')
data
```

```python
X = data.drop(columns='target')
y = data['target']
```

```python
print(X.shape)
print(y.shape)
```

## train test split 

```python

X_train, X_test, y_train, y_test = train_test_split(X,y,
                                                    test_size=0.2,
                                                    random_state=2
                                                    )
```

## fit model 

```python
logistic_regresion_model = LogisticRegression()
logistic_regresion_model.fit(X_train,y_train)
```

```python
decesion_tree_model = DecisionTreeClassifier()
decesion_tree_model.fit(X_train,y_train)
```

```python
# now do predicaction on each model 
y_pred1 = logistic_regresion_model.predict(X_test)
y_pred2 = decesion_tree_model.predict(X_test)
```

```python
pd.DataFrame(data={
    'actual_label': y_test,
    'Logistic Regression Predication' : y_pred1,
    'Decesion Tree Predication' : y_pred2,
})
```

# Accuray
$\text{Accuracy} = \frac{\text{Number of correct Predication}}{\text{Total Number of Predications}}$


```python
from sklearn.metrics import accuracy_score,confusion_matrix
print("Accuracy of Logistic Regression",accuracy_score(y_test,y_pred1))
print("Accuracy of Decision Trees",accuracy_score(y_test,y_pred2))
```

So Decesion Tree perform better

# Confusion matrix 

## binary class

```python
# Logistic Regression
confusion_matrix(y_test,y_pred1)
```

```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
from sklearn.datasets import load_breast_cancer
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,
                                                test_size=0.2,
                                                random_state=2
                                                )
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
logistic_regresion_model = LogisticRegression()
decesion_tree_model= DecisionTreeClassifier()
```

```python
logistic_regresion_model.fit(X_train,y_train)
decesion_tree_model.fit(X_train,y_train)
```

```python
y_pred1 = logistic_regresion_model.predict(X_test)
y_pred2 = decesion_tree_model.predict(X_test)
```

```python

from sklearn.metrics import accuracy_score,confusion_matrix
print("Accuracy of Logistic Regression",accuracy_score(y_test,y_pred1))
print("Accuracy of Decision Trees",accuracy_score(y_test,y_pred2))
```

```python
print('logistic_regresion_model confusion matrix\n')
pd.DataFrame(confusion_matrix(y_test,y_pred1))
```

```python
print('decesion_tree_model confusion matrix\n')
pd.DataFrame(confusion_matrix(y_test,y_pred2))
```
