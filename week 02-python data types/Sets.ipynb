# Set Operation 

## Union ( | ) or 

```python
set_1 = {1,2,3,4,5}
set_2 = {4,5,6,7,8}
```

```python
# dono set ma sa combined item ek ek bar 
set_1.union(set_2)
```

```python
set_1 | set_2
```

## intersection (&) and 

```python
set_1 = {1,2,3,4,5}
set_2 = {4,5,6,7,8}
```

```python
#dono set ma sa Common  item ek ek bar 
set_1.intersection(set_2)
```

```python
set_1 & set_2
```

## negation (~)


```python
# opposite /inverse/ reverse 
set_1.isdisjoint(set_2)
```

## difference ( - )

```python
set_1.difference(set_2)
# vo element jo set_1 ma ha set_2 ma nhi ha   # only set_1
# A -(A into B)
set_1 - set_2
```

```python
set_2.difference(set_1)
#vo element jo set_2 ma ha set_1 ma nhi ha 
set_2 - set_1
```

## symmertic diffrernce( ^ )

```python
set_1.symmetric_difference(set_2) # common element ata do
```
