###`Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`




````python
largest = -1 #[9,34,12,5,54,34]
for the_num in [2,4,6,10,1]:
    if the_num >= largest:
        largest = the_num # smart technique that 
    print(largest, the_num)
# setting a number largest 
# then loop through the list 
# fir check karo yadi jo num list ma vo greatar ha ki nhi hamara largest 
# variable sa agar vo largest ha tho then largest ko vahi value assign karo
# or pura loop ma value update kar do 
````

````python
ls = [2,4,6,10,1]
result = []
sum = 0
for first_loop in ls:
    for second_loop in ls:
        if first_loop<=second_loop:
            sum = sum+ second_loop
        
    result.append(sum)

print(result)
````

###`Problem 6:` Find list of common unique items from two list. and show in increasing order

`Input`

```
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

`Output:`
```
[34, 67, 89]
```


````python
#common element find karna ha 
# list 1 ma loop
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
common_element = []
for i in num1:
    if i in num2:
        common_element.append(i)
print(sorted(common_element))
````

````python
# use concept of interction find commnon unique element
#convert list to list 
set_1 = set(num1)
set_2 = set(num2)
# do intersection operation and convert into list
# sorted function ma internally hi list ma convert kar diya 
sorted(set_1.intersection(set_2))
````

*###`Problem 7:` Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.

`Input:`
```
['1ac21', '23fg', '456', '098d','1','kls']
```

`Output:`
```
['456', '23fg', '1ac21', '1', 'kls', '098d']*

```

````python
# make a loop then compare the previous value to this value 
# tho muja alpha numeric string ma sa numeric value bhar nikalni ha
# fir product karna unko list ma karnha then use sort 
List = ['1ac21', '23fg', '456', '098d','1','kls']
for item in List:
    print(sorted(item))
    # for character in sorted(item) :
    #     print(character)
````

````python
List = ['1ac21', '23fg', '456', '098d', '1', 'kls']
product_value = []

for item in List:
    product = 1
    for character in item:  # Iterate over the characters of the current item
        if character.isdigit():
            product = product * int(character)
            
    product_value.append(product)
    
print(product_value)
# now use list comprenshison 
list(zip(product_value,List))
print(sorted(list(zip(product_value,List)),reverse=True))
````

### `Problem 10:` Add Space between Potential Words.

**Example:**

Input:

```bash
['campusxIs', 'bestFor', 'dataScientist']
```

Output:
```bash
['campusx Is', 'best For', 'data Scientist']
```

````python
ls = ['campusxIs', 'bestFor', 'dataScientist']
# list ka andr loop chalo when you find captital letter use split func
empyty_ls = []
for item in ls:
    for character in item:
        if character.isupper():
            pass
            empyty_ls.append(character.split())
print(empyty_ls)
#[i for i in ls j]
````

````python
test_list = ['campusxIs', 'bestFor', 'dataScientist']
res = []
for i in test_list:
    #print(i)
    temp = [[]]
#break the character inot list if you find upper 
    for character in i:
        #print(character)
        if character.isupper():
            temp.append([])
        
        temp[-1].append(character)
    #print(temp)
    temp_string = ""
    for item in temp: # make them in a word of list use join 
        #print("".join(item))
        temp_string = temp_string + "".join(item) + " "
    res.append(temp_string[0:-1]) # last space remove karna ka liya 
print(res)
````


### `Problem 15:` Write a list comprehension that can flatten a nested list

Input<br>
matrix = [<br>
  [1,2,3],<br>
          [4,5,6],<br>
          [7,8,9]<br>]<br>

Output:<br>
[1, 2, 3, 4, 5, 6, 7, 8, 9]

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

flattened_list = [element for row in matrix for element in row]

#element for row in matrix  -> outer loop in each row
#for element in row - > ya har list ka andr ja raha ha 
print(flattened_list)
````

````python
flattened_list = []

for row in matrix:
    for element in row:
        flattened_list.append(element)

print(flattened_list)
````

````python
flattened_list = []

for first_loop in matrix:
    for second_loop in first_loop:
        flattened_list.append(second_loop)

print(flattened_list)
````
