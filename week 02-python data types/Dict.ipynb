# Dictionary
Dictionary in python is  a collection of keys,values pair  used to store data values like a map 
which unlike another data type store  which hold a  single value  as an a element  

````python
# structure 
# dict = { key: value   , key: value  ,  key: value  }

# read able good written code 
# dict = {
#         key: value  ,
#         key: value  ,   
#         key: value   }
````

````python
List = [10, 20, 30, 40]
print(List)
Dictionary = {
    "string": "focus",
    "int": 20 ,
    "float": 3.14 ,
    "bolllean": True
}
print(Dictionary)
# list ka index key ka  work kar raha ha 
# in dict any data - type allowed 
````
