 # high school promblem
 With your high school reunion fast approaching, you decide to get in shape and lose some weight . You record your weight every day for five weeks starting on a Monday.

```python
import numpy as np

dailywts = 185 - np.arange(5*7)/5

print(dailywts)
```

Given these daily weights, build an array with your average weight per weekend1.

```python
dailywts.size # or
# tho total 35 days mena apna weight record kiya 
# starting day --> monday
# ending day --> sunday

# 1 weeek = 7 days 
# x week = 35 days 
# 5 weeks
# aor mera mission ha har ek week ma avg weight kasa karga
```

```python
# dailywts array 1d ha uss 1 into 7 matrix bano 
weekly_dailywts = dailywts.reshape(5,7)
print(weekly_dailywts)
# fir har ek row ka avg la lo
```

```python
avgweight= np.mean(weekly_dailywts,axis=0)
for i in avgweight:
    print("weight on week --> ", i)
```

# Goldminer
After binge watching the discovery channel, you ditch your job as a trial lawyer to become a gold miner  . You decide to prospect five locations underneath a 7x7 grid of land. How much gold do you uncover at each location?

```python
import numpy as np
np.random.seed(5555)
gold = np.random.randint(low=0, high=10, size=(7,7))
print(gold)
# how much gold in each location 
```

```python
# location coordinate of gold ex first 0,4 --> 2
locs = np.array([
    [0,4],
    [2,2],
    [2,3],
    [5,1],
    [6,3]
])
print(locs)
```

## notes

- locs states the coordinates of the five locations where you dig
