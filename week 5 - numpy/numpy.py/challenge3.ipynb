```python
import numpy as np

generator = np.random.default_rng(123)
ratings = np.round(generator.uniform(low=0.0, high=10.0, size=(10, 2)))
ratings[[1,2,7,9], [0,0,0,0]] = np.nan

print(ratings)
```

```python
movie= ratings[:,0]
movie[[1,2,7,9]]= 0
# index = np.isnan(movie)
# np.argwhere(index)
movie
```

```python
director_rating = ratings[:,1]
director_rating
```

```python
x = (movie + director_rating)[:,np.newaxis]
x
```

```python
np.concatenate((ratings,x), axis=1)
```
