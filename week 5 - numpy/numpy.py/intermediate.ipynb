##  Broadcasting

1. (m,n) +(1,n) = (m,n)
2. (1,n) +(m,n) = (m,n)
3. (1,n) +(1,m) = (m,n)

1. Moving backwards from the last dimension of each array,
2. We check if their dimensions are "compatible". Dimensions are compatible if they are equal or either of them is 1
3. If all of A's dimensions are compatible with B's dimensions, or vice versa, they are compatible arrays

```python
import numpy as np
arr1 = np.arange(0,5).reshape(1,5)
arr2 = np.arange(0,5).reshape(5,1)
print(arr1 + arr2)
```

# Boolean indexing

With boolean indexing, you can subset an array A using another array B of boolean values.

```python
import numpy as np
foo = np.array([
    [3, 9, 7],
    [2, 0, 3],
    [3, 3, 1]
])
```

```python
#foo ==3
#both are same to do subset
mask = foo ==3
print(mask)
```

```python
#foo[foo ==3]
print(foo[mask])
```

```python
foo[mask] = 0
print(foo)
```

```python
rows_1_and_3 = np.array([True,False,True])
columns_1_and_2 =  np.array([False,True,True])
```

```python
foo[rows_1_and_3,:] #like foo[[0,2],:] object[[row_index,col_index]]
```

```python
foo[:,columns_1_and_2 ] #like foo[:,[1,2]]
```

# Logical operators

```python
b1 = np.array([False, False, True, True])
b2 = np.array([False, True, False, True])

b1 & b2  # [False, False, False,  True], and, intersection
b1 | b2  # [False,  True,  True,  True], or, union
b1 ^ b2  # [False,  True,  True, False], xor
~np.array([False, True]) # array([ True, False]) inverse
```

# Nan
You can use NaN to represent missing or invalid values. NaN is a floating point constant that numpy reserves and treats specially.

```python
bot = np.ones(shape=(3,4))
bot[[0,2],[1,2]] = np.nan
print(bot)
print(np.isnan(bot))
```

Ask question is there a null value 

```python
bot==np.nan
#observe all are false 
#logic what is empty ==empty
#operation are notdefined
```

```python
#all true
#nothing same beahaviour
# empty means empty 
bot != np.nan
```

```python
#this is working
print(np.isnan(bot))
```

NaN is a special floating point constant, so it can only exist inside an array of floats. If you try inserting NaN into an array of integers, booleans, or strings, you’ll get an error or unexpected behavior.
