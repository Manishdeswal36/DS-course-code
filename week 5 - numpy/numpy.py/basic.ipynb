# why numpy exists 

### in native python store data

```python
# suppose you have a lot of data of price of stock measured every second
# for a year that's about 32 million values
# for native python you reprsent this information 
# data structure  in list. lets see example

prices = []
N =60*60*60*365
# N = secoonds * minute * hour * year
for i in range(N):
    prices.append(100 + i/100)

print(prices[:5]) # print first five price outputs
```

### cal avg of prices in native python

```python
avg = 0.0
for p in prices:
    avg += p/len(prices)

print(avg)
```

### store data in numpy

```python
# this is slow in native python lets do in numpy 

import numpy as np

prices = 100 + np.array(N)/100
print(prices[:5])
```

### avg in numpy

```python
avg = np.mean(prices)
print(avg)

# all the math operations
```

# Rules For Every NumPy Array¶
There are two basic rules for every numpy array..

1. Every element in the array must be of the same type and size.
2. If an array's elements are also arrays, those inner arrays must have the same type and number of elements as each other. In other words, multidimensional arrays must be rectangular, not jagged.

## hetergenous array

```python
# what happen if i create a hetergenous array
import numpy as np
mixed_array = np.array(['hello',2, 3.14, True])
print(mixed_array)
# it convert every data -type to string interally
mixed_array.dtype
# <U32 = unicode characters store's 32 bits only 
```

```python
# if you store a very large sting compare to 32 bits only
mixed_array[0] = 'a really really really large string '
```
