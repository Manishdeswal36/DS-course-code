# np.where

### find all indices with value greator than 50 

```python
import numpy as np 
array = np.random.randint(1,100,15)
array
```

```python
np.where(array>50) # you can fetch the index position where condition is true
```

```python
#np.where(condition,True,False)
```

# np.corrcoef

```python
# correlation coefficient nikalan ka liya 
# 1 +ve  = both inc or dec one direction
# 0  = no relation 
# -1  -ve = one inc or another vari dec  
salary = np.array([20000,40000,25000,35000,60000])
experience = np.array([1,3,2,4,2])

np.corrcoef(salary,experience)
```

|            | salary | experience
| -----------|--------|-----------
| salary     |  1     |   0.25
| experience | 0.25   |     1

```python
# salary ka salary ka sath coefficient correlation 1 hoga
# ek Quanity ka apna sath correlation tho 1 hi hoga
# salary ka experinece ka sath 0.25 correlation ha 
```

## np.union1d

```python
array_1 = np.array([1,2,3,4,5])
array_2 = np.array([3,4,5,6,7])
np.union1d(array_1,array_2)
```

## np.intersect1d

## np.setdiff1d


```python
# items A ma ha B ma nhi ha 
np.setdiff1d(array_1,array_2)
```

```python
# items B ma ha A ma nhi ha
np.setdiff1d(array_2,array_1)
```


## np.setxor1d
