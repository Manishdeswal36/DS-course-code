# method 1  to plot figure

```python
fig = plt.figure()
ax = fig.add_subplot(111)# 1 row 1 column 1figure number
ax.set(
    xlim = [0.5,4.5],
    ylim = [-2,8],
    title = 'an example axes',
    xlabel = 'X-axis',
    ylabel = 'Y-axis'
)
plt.show()
```

# method 2 plot figure 

```python
fig = plt.figure()
ax = plt.figure().add_subplot()
ax.set_ylim(1,5)
ax.set_xlim(10,40)
ax.set_xlabel('X- axis')
ax.set_ylabel('Y-axis')
ax.set_title('Awesome')
plt.show()
```
