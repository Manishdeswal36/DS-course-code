# Stacking 

```python
# stacking and splitting aap dono case ma array ma changes kar raha ho 
# aap 2 ya 2 sa jada numpy array ko jod sakta ho
# 2 tarika ha 
# ya tho aap horizontal jodo side wise  left and right 
# ya tho app vertical jodo  uppar nicha side down 
```

```python
# example we have two numpy array 
# [  2 by 2  ]  [ 2 by 2 ]
# you two stack by horizontal and vertical 
# horizontal   [ 2 by 4      ] rows same column increase 

# vertical   [ 4 by 2] rows increase    column same 

```

### horizontal 

```python
import numpy as np 
array_1 = np.arange(0,4).reshape(2,2)
array_2 = np.arange(4,8).reshape(2,2)
# print(array_1)
# print(array_2)
#np.hstack(  (pass in tuple )  )
np.hstack( (array_1,array_2))
```

### vertical 

```python
np.vstack( (array_1,array_2))
```

# splilting 

## horizontal

```python
matrix = np.arange(0,8).reshape(2,4)
matrix
```

## vertical 

```python
matrix = np.arange(0,8).reshape(2,4)
matrix
```
