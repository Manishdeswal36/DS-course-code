###`Q-1:` Create a random 3x4 matrix with value between 0-100. And perform below tasks

    i. Sort this matrix. np.sort()
    ii. Sort this matrix based on values in 2nd column.
    iii. Sort this matrix based on max value in each row.
    iv. Sort based on elements value.


See examples:
```
arr =
    [[92 90 74]
    [ 6 63 93]
    [15 93 96]
    [70 60 48]]
```
```
i. np.sort
[[74 90 92]
 [ 6 63 93]

 [48 60 70]]

ii. based on 2nd column
[[70 60 48]
 [ 6 63 93]
 [92 90 74]
 [15 93 96]]

iii. based on row max- ascending
[[15 93 96]

 [70 60 48]]

iv. based on elements value
[[ 6 15 48]
 [60 63 70]
 [74 90 92]
```

````python
import numpy as np
np.random.seed(1235)
random_array = np.random.randint(low=1,high=100 ,size = 12).reshape(3,4)
random_array
````

````python
# Question 1 
np.sort(random_array)# default sort last axis yani 1 hota yani rows wise
````

````python
# Question 1
np.sort(random_array,axis=0) # coloumn wise 
````

````python
# question 2 
# second column find the index using argsort 
random_array[:,1].argsort()
# usma dalo # now sort by second column 
random_array[random_array[:,1].argsort()]
````

````python
# Question 3 
# their is not any function in numpy we use python func
np.array(sorted(random_array,key=lambda x:max(x)))
````

````python
# Question 4
# when axis = none it first flat the array than sorted ascending order 
np.sort(random_array,axis=None).reshape(3,4)
````

###`Q-2:` There is an array of marks of 5 students in 4 subjects. Further you are asked to perform below task.
    i. Add marks every student of an extra subject in the same array.
    ii. Add two new students marks in respective 5 subjects.(one subject added in above task)
    iii. Add extra column with sum of all subjects(5-subjects) marks
    iv. Sort the array(non-ascending order) on total marks column--one added in above task. Show top 2 rows.

Note: Change dimension of arrays during concatenation or appending if required.

Given Array-
```
marks = [[13, 10,  9, 33],
       [63, 46, 90, 42],
       [39, 76, 13, 29],
       [82,  9, 29, 78],
       [67, 61, 59, 36]]

extra_subject = [41, 87, 72, 36, 92]
#Two extra students record-
rec1 = [77, 83, 98, 95, 89]
rec2 = [92, 71, 52, 61, 53]
```

i. Add marks every student of an extra subject in the same array.
    

````python
# subject y axis coloumn axis = 0
# marks  x axis rows axis = 1
````

````python
marks = np.array([
        [13, 10,  9, 33],
        [63, 46, 90, 42],
        [39, 76, 13, 29],
        [82,  9, 29, 78],
        [67, 61, 59, 36]
        ])
extra_subject = np.array([41, 87, 72, 36, 92])
extra_subject = np.expand_dims(extra_subject,axis=1)
print(extra_subject) # we shape the array 

marks = np.concatenate((marks,extra_subject),axis=1)
marks
````

ii. Add two new students marks in respective 5 subjects.(one subject added in above task)


````python
rec1 = np.array([77, 83, 98, 95, 89]).reshape(1,-1)
rec1
````

````python
# make the correct shape of rec1 and rec2
rec1 = np.array([77, 83, 98, 95, 89]).reshape(1,-1)
rec2 = np.array([92, 71, 52, 61, 53]).reshape(1,-1)
marks = np.concatenate((marks,rec1,rec2),axis=0) # add the record 
````

````python
marks
````

````python
# sum of each student subject(left to right )
sum_of_std= np.sum(marks,axis= 1,keepdims = True )
sum_of_std
````

````python
marks = np.concatenate((marks,sum_of_std),axis=1) # add the sum of each student subject
````

````python
marks
````

iii. Add extra column with sum of all subjects(5-subjects) marks


iv. Sort the array(non-ascending order) on total marks column--one added in above task. Show top 2 rows.

````python
np.array(sorted(marks,key=lambda x:x[-1],reverse=True)) # sort by sum of each student subject
````

### `Q-7:` You are given a array. You have to find the minimum and maximum array element and remove that from the array.

```python
import numpy as np

np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))
```

````python
import numpy as np

np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))
# find max and min indices using argmax and argmin 
print(arr)
max_element = arr.argmax()
min_element = arr.argmin()
print(arr.max(),arr.min())
arr = np.delete(arr,max_element,axis=1)
arr = np.delete(arr,min_element,axis=1)

print(arr.max(),arr.min())
````

### `Q-9:` You are given a array ($arr ∈ [0, 1]$). First you have round off the elements upto 3 decimal places and compare that

````python
array = np.random.uniform(1,100,30)
array = np.round(array,3) # round the array to 3 decimal places
print("0th perncntile minimimu value of arry :",np.percentile(array,0))
print("100th perncntile minimimu value of arry :",np.percentile(array,100))
print('difference b/w 51 and 50 perceniile :',np.percentile(array,51)-np.percentile(array,50))
````
