### `Q-6`: Write a python function which infinitely prints natural numbers in a single line. Raise the **StopIteration** exception after displaying first 20 numnbers to exit from the program.

````python
num = 0          # start the loop --> intial conditon
# but me this true to make infinte loop 
while num > -1:   # stop  the loop --> by making false conditon 
    if num == 20:
        raise StopIteration

    num = num +1   # continue the loop -->permform operation by making conditon true
    print(num)    # display on screen 
````
