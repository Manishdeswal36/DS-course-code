```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Ridge,Lasso ,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train ,X_test ,y_train,y_test = train_test_split(X,y,
                                                test_size=0.2,
                                                random_state=2
                                                )
```

```python
# Linear Regression
linear_regression_model = LinearRegression()
linear_regression_model.fit(X_train,y_train)

y_pred = linear_regression_model.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# ridge Regression
ridge_regression_model = Ridge(alpha=0.1)
ridge_regression_model.fit(X_train,y_train)

y_pred = ridge_regression_model.predict(X_test)
r2_score(y_test,y_pred)
```

```python
lasso_regression_model = Lasso(alpha=0.01)
lasso_regression_model.fit(X_train,y_train)

y_pred = lasso_regression_model.predict(X_test)
r2_score(y_test,y_pred)
```

```python
elasticnet_regression_model = ElasticNet(alpha=0.005,l1_ratio=0.9)
elasticnet_regression_model.fit(X_train,y_train)

y_pred = elasticnet_regression_model.predict(X_test)
r2_score(y_test,y_pred)
```

Grid search yani hyperparameter tuning
