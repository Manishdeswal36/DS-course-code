### super sa method call

````python
class  Phone:

    def __init__(self,price, brand, camera):
        print("inside phone constructer")
        self.price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("buying a phone  parent class")
        
class Smartphone(Phone):
    
    def buy(self):
        print("buying a smartphone child class")
        super().buy() # syntax for parent class ka method ko bhi call karna ka 

smart_phn_object= Smartphone(1000,"apple", 4)
smart_phn_object.buy()
````

### super sa constructor call
how to use super keyworad smartly and optimise code
- child ka object create only call child constuctor
- parent ka object create only call parent constuctor.
- what if i use or call t to `both constructor` solution super . 
- example price brand camera har phone ma hota ha
- but operating system (ios,andriod)and ram(2,4,6,8)gb vary for every 

````python
class  Phone:

    def __init__(self,price, brand, camera):
        print('super sa 3 variable intiliase parent')
        self.price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("buying a phone  parent class")
        
class Smartphone(Phone):

    def __init__(self, price, brand, camera,os, ram ):
        print('first go to smartphone constuctor 5 variable intisliase then super')
        super().__init__(price, brand, camera)# mana super sa kaha constructaor ka ya input parent ma
        self.os = os             # la lo ab vo parent  construct ma jay aga
        self.ram = ram 
        print('then intislase remaing 2 varialbe ')                                       

child_object =  Smartphone(20000,"samsung", 14,'anrdroid', 4)
````

````python
from abc import ABC, abstractmethod

class BankApp(ABC):

    def database(self):
        print('conncet to database ')

    @abstractmethod
    def security(self):
        print("first implement security ")
````

````python
class MobileApp(BankApp):
    
    def mobile_login(self):
        print('mobile login page')
````

````python
mob = MobileApp() # type: ignore
#TypeError: Can't instantiate abstract class MobileApp with abstract method security

# matlab huma security method dalna hoga
# aap database sa tab tak connect nhi hoga jak tak aap apni
# class ka andr security class implement nhi karoga
````

````python
class MobileApp(BankApp):
    
    def mobile_login(self):
        print('mobile login page')

    def security(self):
        print('security work')
# create mobileapp object
mob = MobileApp()
mob.database() # now it connect to db after implementing security class
# now it can create object of child  
````
