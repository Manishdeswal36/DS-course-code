```python
import numpy as np
import pandas as pd

import plotly.graph_objects as go
import plotly.express as px
```

```python
latitude_longitude = pd.read_csv("district wise centroids.csv")
latitude_longitude
```

```python
# in 2011 their are 640 district in india but our data have 594 distict so not much data is missing . 
# lets check if their is duplicated data or not
latitude_longitude.isnull().sum()
```

```python
censes = pd.read_csv('india-districts-census-2011.csv')
censes
```

```python
censes.info()
```

```python
# so their are 116 numericals coloumns and 2 cateogrical coloumns
# ek district code ko hamna drop kiya ha 
```

```python
censes.columns.to_list()
```

```python
censes_df = censes[['District name', 'Population', 'Male','Female','Literate','Male_Literate', 'Female_Literate','Housholds_with_Electric_Lighting',
'Households_with_Internet','Rural_Households',"Urban_Households",'Total_Power_Parity' ,]]
censes_df
```

```python
final_df = latitude_longitude.merge(censes_df,left_on='District',right_on='District name').drop(columns='District name')
final_df
```

```python
# we are adding two more coloumns literacy rate and sex ratio
final_df['sex_ratio'] = round((final_df['Female']/final_df['Male'])* 100)
```

```python
final_df['literacy_rate'] = round((final_df['Literate']/final_df['Population']) * 100)
```
