# st.cache decorator 

```python
# so whenver you click on rerun so its rerun the entire file 
# again and again 
# it is a small file does not have much effect 
# but if their is multiples files then use
# streamlit cache data decoratoar 
@st.cache_data # type: ignore
```

```python
import pandas as pd
import streamlit as st


st.title("📞What are the usual failing products ?")

@st.cache_data
def load_data(path: str):
    data = pd.read_csv(path)
    return data

df = load_data(path='./data/Financial Consumer Complaints.csv')
st.dataframe(df,hide_index=True)
```

```python
import pandas as pd
import streamlit as st


st.title("📞What are the usual failing products?")
# it will give you the recent 5 cache entries 
@st.cache_data(max_entries=5)
def load_data(path: str):
    data = pd.read_csv(path)
    return data

df = load_data(path='./data/Financial Consumer Complaints.csv')
st.dataframe(df,hide_index=True)
```

```python
import pandas as pd
import streamlit as st


st.title("📞What are the usual failing products?")

@st.cache_data(max_entries=5)
def load_data(path: str):
    data = pd.read_csv(path,parse_dates=['Date Sumbited',"Date Received"])
    return data

df = load_data(path='./data/Financial Consumer Complaints.csv')


all_products = df['Product'].unique()

selected_product = st.selectbox(label='All Products',options=all_products)#options=None)

st.title(selected_product)

st.dataframe(df,hide_index=True)

```

# setup 

```python
import pandas as pd
import streamlit as st


########################################
# Setup
########################################

st.set_page_config(
    page_title='Complaning Dashbord',
    page_icon='📈'
)

########################################
# load data 
########################################
st.title("📞What are the usual failing products?")

@st.cache_data(max_entries=5)
def load_data(path: str):
    data = pd.read_csv(path,parse_dates=['Date Sumbited',"Date Received"])
    return data

df = load_data(path='./data/Financial Consumer Complaints.csv')
all_products = df['Product'].unique()


########################################
# UI 
########################################



selected_product = st.selectbox(label='All Products',options=all_products)#options=None)

st.title(selected_product)


st.dataframe(df,hide_index=True)

```
