### **Problem-3:** Make a 3D *Surface* plot of this below mathematical equation.

$$z = |x| - |y|$$

````python
x = np.linspace(-10,10,50)
y = np.linspace(-10,10,50)
xx, yy = np.meshgrid(x,y)
# our function equation 
z = np.abs(xx) - np.abs(yy)
````

````python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection = '3d')

ax.plot_surface(xx,yy,z ,cmap = 'viridis')
````

````python
# lets also make in plotly now it is interactive 

import plotly.graph_objects as go

trace = go.Surface(x=x,y=y,z=z)
data = [trace]

layout = go.Layout(title='3d surface plot')
fig = go.Figure(data,layout)
fig.show()
````

### `Problem 9:` Create a surface plot and the 2 types of the contour plots of the below equation.

$$z = sin(\sqrt{x^2 + y^2})$$


````python
x = np.linspace(-10,10,50)
y = np.linspace(-10,10,50)
xx, yy = np.meshgrid(x,y)
z = np.sin(np.sqrt(xx**2 + yy**2))


fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection = '3d')

ax.plot_surface(xx,yy,z ,cmap = 'viridis')

````

````python
fig = plt.figure(figsize=(12,8))
ax = plt.subplot()
p = ax.contour(xx,yy,z,cmap = 'viridis')
fig.colorbar(p)
````

````python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()
p = ax.contourf(x,y,z,cmap = 'viridis')
fig.colorbar(p)
````

### `Problem 10:` Create a surface plot and the 2 types of the contour plots of the below equation.

$$z = tan(\log_2({x^2 + y^2})$$

````python
x = np.linspace(-10,10,50)
y = np.linspace(-10,10,50)
xx, yy = np.meshgrid(x,y)
z = np.tan( np.log2( (xx**2 + yy**2) ) )

fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection = '3d')
ax.plot_surface(xx,yy,z)
````

````python
fig = plt.figure(figsize=(12,8))
ax = plt.subplot()
p = ax.contour(xx,yy,z)
fig.colorbar(p)
````

````python
fig = plt.figure(figsize=(12,8))
ax = plt.subplot()
p = ax.contourf(x,y,z, cmap = 'viridis')
fig.colorbar(p)
````
