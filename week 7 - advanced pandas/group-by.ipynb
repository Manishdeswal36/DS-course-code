# aggreate func
Pandas offers a variety of aggregation functions that can be used with the `.agg()` method. Here's a list of common ones:

- **Mean**: Calculates the mean of the values.
- **Median**: Finds the median of the values.
- **Mode**: Returns the most frequently occurring value.
- **Sum**: Sums up all the values.
- **Min**: Finds the minimum value.
- **Max**: Finds the maximum value.
- **Std**: Calculates the standard deviation.
- **Var**: Calculates the variance.
- **Count**: Counts the number of non-null values.
- **Prod**: Calculates the product of all values.

You can also use custom functions or apply different functions to different columns by passing a dictionary to `.agg()`

## find normalized IMDB rating group wise

$ normalised-score = \frac{x - x_{min}}{x_{max}-x_{min}}$

````python
# aap har ek group ka basis par analysis karna chata ho 
# example india level par exam hua aap har state ma kitna rank aya bacha ka ya dekna chata ho 
````

````python
def normal(group):
    # apply the normalising formula 
    group['normal_rating'] = round((group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max()- group['IMDB_Rating'].min()),2)
    return group
````

````python
genres.apply(normal)
````

### Create a function that can return the highest score of any batsman


````python
temp_df
````

````python
# filter the rows by batsman
temp_df = ipl[ipl['batsman'] =='V Kohli']
temp_df.groupby('(match_id')['batsman_runs'].sum().sort_values(ascending=False).values[0]
````

````python
def highest_score(batsman):
    temp_df = ipl[ipl['batsman'] == batsman]
    return temp_df.groupby('(match_id')['batsman_runs'].sum().sort_values(ascending=False).values[0]
````

# Notes

Both **Pandas joins** and **SQL joins** are based on the principles of **set theory**, particularly operations on sets such as **union, intersection, and difference**. Let’s break this down in terms of **set theory**:

---

### **1️⃣ INNER JOIN → Intersection (A ∩ B)**
- **Concept**: Returns only the matching rows between two datasets.
- **Set Theory Equivalent**: The **intersection** of two sets.
- **Example**:
  - SQL: `SELECT * FROM A INNER JOIN B ON A.id = B.id`
  - Pandas: `df1.merge(df2, on="id", how="inner")`
  
#### **Set Notation**:
\[
A \cap B = \{ x \mid x \in A \text{ and } x \in B \}
\]

#### **Venn Diagram**:
```
     +---------+
     |   A ∩ B  |  <-- Only common elements
     +---------+
```

---

### **2️⃣ LEFT JOIN → Left Set with Intersection (A ∪ (A ∩ B))**
- **Concept**: Returns all rows from the left dataset and only matching rows from the right dataset.
- **Set Theory Equivalent**: The **union of set A and the intersection (A ∩ B)**.
- **Example**:
  - SQL: `SELECT * FROM A LEFT JOIN B ON A.id = B.id`
  - Pandas: `df1.merge(df2, on="id", how="left")`

#### **Set Notation**:
\[
A \cup (A \cap B) = A
\]

#### **Venn Diagram**:
```
    +----------+
    |    A     | <-- Includes all of A
    |  A ∩ B   | <-- Matching elements from B
    +----------+
```

---

### **3️⃣ RIGHT JOIN → Right Set with Intersection (B ∪ (A ∩ B))**
- **Concept**: Returns all rows from the right dataset and only matching rows from the left dataset.
- **Set Theory Equivalent**: The **union of set B and the intersection (A ∩ B)**.
- **Example**:
  - SQL: `SELECT * FROM A RIGHT JOIN B ON A.id = B.id`
  - Pandas: `df1.merge(df2, on="id", how="right")`

#### **Set Notation**:
\[
B \cup (A \cap B) = B
\]

#### **Venn Diagram**:
```
    +----------+
    |    B     | <-- Includes all of B
    |  A ∩ B   | <-- Matching elements from A
    +----------+
```

---

### **4️⃣ FULL OUTER JOIN → Union (A ∪ B)**
- **Concept**: Returns all rows from both datasets, with `NULLs` where no match is found.
- **Set Theory Equivalent**: The **union of both sets**.
- **Example**:
  - SQL: `SELECT * FROM A FULL OUTER JOIN B ON A.id = B.id`
  - Pandas: `df1.merge(df2, on="id", how="outer")`

#### **Set Notation**:
\[
A \cup B = \{ x \mid x \in A \text{ or } x \in B \}
\]

#### **Venn Diagram**:
```
    +---------+
    |  A ∪ B  | <-- Includes all elements from A and B
    +---------+
```

---

### **5️⃣ LEFT ANTI JOIN → A - (A ∩ B)**
- **Concept**: Returns rows from the left dataset that **do not** have a match in the right dataset.
- **Set Theory Equivalent**: **Set difference**.
- **Example**:
  - SQL: `SELECT * FROM A WHERE A.id NOT IN (SELECT id FROM B)`
  - Pandas: `df1.merge(df2, on="id", how="left", indicator=True).query('_merge == "left_only"').drop('_merge', axis=1)`

#### **Set Notation**:
\[
A - B = \{ x \mid x \in A \text{ and } x \notin B \}
\]

#### **Venn Diagram**:
```
    +------+
    |  A   | <-- Only elements in A, not in B
    +------+
```

---

### **6️⃣ RIGHT ANTI JOIN → B - (A ∩ B)**
- **Concept**: Returns rows from the right dataset that **do not** have a match in the left dataset.
- **Set Theory Equivalent**: **Set difference**.
- **Example**:
  - SQL: `SELECT * FROM B WHERE B.id NOT IN (SELECT id FROM A)`
  - Pandas: `df1.merge(df2, on="id", how="right", indicator=True).query('_merge == "right_only"').drop('_merge', axis=1)`

#### **Set Notation**:
\[
B - A = \{ x \mid x \in B \text{ and } x \notin A \}
\]

#### **Venn Diagram**:
```
    +------+
    |  B   | <-- Only elements in B, not in A
    +------+
```

---

### **Final Comparison Table**
| Join Type | SQL Query | Pandas Syntax | Set Theory Equivalent |
|-----------|----------|--------------|------------------------|
| **INNER JOIN** | `SELECT * FROM A INNER JOIN B ON A.id = B.id` | `df1.merge(df2, on="id", how="inner")` | **Intersection (A ∩ B)** |
| **LEFT JOIN** | `SELECT * FROM A LEFT JOIN B ON A.id = B.id` | `df1.merge(df2, on="id", how="left")` | **A with intersection (A ∪ (A ∩ B))** |
| **RIGHT JOIN** | `SELECT * FROM A RIGHT JOIN B ON A.id = B.id` | `df1.merge(df2, on="id", how="right")` | **B with intersection (B ∪ (A ∩ B))** |
| **FULL OUTER JOIN** | `SELECT * FROM A FULL OUTER JOIN B ON A.id = B.id` | `df1.merge(df2, on="id", how="outer")` | **Union (A ∪ B)** |
| **LEFT ANTI JOIN** | `SELECT * FROM A WHERE A.id NOT IN (SELECT id FROM B)` | `df1.merge(df2, on="id", how="left", indicator=True).query('_merge == "left_only"')` | **Set difference (A - B)** |
| **RIGHT ANTI JOIN** | `SELECT * FROM B WHERE B.id NOT IN (SELECT id FROM A)` | `df1.merge(df2, on="id", how="right", indicator=True).query('_merge == "right_only"')` | **Set difference (B - A)** |

---

### **Summary**
- SQL and Pandas **joins** are fundamentally based on **set operations**.
- **INNER JOIN** is equivalent to **intersection (A ∩ B)**.
- **LEFT JOIN** and **RIGHT JOIN** keep one full set and its intersection with the other.
- **FULL OUTER JOIN** is equivalent to **union (A ∪ B)**.
- **LEFT ANTI JOIN** and **RIGHT ANTI JOIN** are **set differences (A - B, B - A)**.
