````python
# hum batsmna ka bana raha ha aap cha ho tho bowler ka bhi bana sakta ho agar apko time mila tho.
````

````python
# haman jo dream 11 website ka unka point methods 
# point dena ka liya har player ko we need three things
# har batsman ka har match ka 3 data ho
# usna kitna runs banaya
# usna kitna choka ( 4's) banaya
# usna kitna  chaka (6's) banaya 
# tho hum usko scoring kar sakta ha .
````

````python
# abhi data ko hum ek intermediate stage ma la kar ayega
# tho usma kiya kiya coloumns hoga
````

- match id
- player name
- run = kitna runs banaya palyer na
- fours = kitna 4's banya 
- sixes = kitna 6's banaya
- balls = us mathc ma player na kitna balls kheli
- captain = us match ma kon captain tha . 

````python
# ek aor level ka upar apko scoring karni ha
# strike rate yadi ek batsman 100 balls khelga tho kitna runs banya ga
# example 100 balls khel kar yadi 120 runs banya tho strike rate is 120
# yadi 100 balls khel kar 200 runs banya tho stike rate hoga 200
# fir waha par rules banya ha
# 170 sa upar stiker rate par +6 points whegra whegra aor bhi 
````

````python
temp_df = player.merge(player_captain,on='Player_Id')[['Player_Name','Match_Id','Is_Captain']]
temp_df
````

````python
delivery = delivery.merge(temp_df, left_on=['ID','batter'],right_on=['Match_Id','Player_Name'],how='left').fillna(0)
````

````python
# deko delivery wala recent data tha year 2021
# player and player captain data humna kahi aor sa liya tha
# tho last wala ma captain ka information nahi ha
# chalo usa 0 sa fill kar deta ha 
````

````python
runs = delivery.groupby(['ID','batter'])['batsman_run'].sum().reset_index()
balls = delivery.groupby(['ID','batter'])['batsman_run'].count().reset_index()
````

````python
runs
# usna kitna runs banaya
````

````python
balls
# usna kitni balls kheli
````

````python
# calculate also the 4's and 6's 
# har batsmane har match ma kitna 4's and 6's banya 
fours = delivery.query('batsman_run == 4').groupby(['ID','batter'])['batsman_run'].count().reset_index()
sixes = delivery.query('batsman_run == 6').groupby(['ID','batter'])['batsman_run'].count().reset_index()
````

````python
fours
````

````python
sixes
````

````python
# sabhi na 4's score nahi kiya isliya rows come ho gaya 
# so join on left .
````

````python
final_df = runs.merge(balls,on=['ID','batter'],suffixes=('_runs','_balls')).merge(fours,on=['ID','batter'],how='left').merge(sixes,on=['ID','batter'],how='left')
````

````python
final_df
````

````python
# replace nan values with 0 
final_df.fillna(0,inplace=True)
````

````python
final_df.rename(columns={
    'batsman_run_runs':'runs',
    'batsman_run_balls':'balls',
    'batsman_run_x':'fours',
    'batsman_run_y':'sixes'
},inplace=True)
````

````python
final_df['sr'] = round((final_df['runs']/final_df['balls'])*100,2)
````

````python
# now merge with captian 
````

````python
final_df = final_df.merge(temp_df,left_on=['ID','batter'],right_on=['Match_Id','Player_Name'],how='left').drop(columns=['Player_Name','Match_Id']).fillna(0)
````

````python
final_df
````

````python
# now ab hum karega scoring 
````

````python
# player ka match ka upar usko dream11 ka score dena ha 
````

````python
def dream11(row):
    score = 0

    score = score + row['runs'] + row['fours'] + 2*row['sixes']

    if row['runs'] >= 100:
        score = score + 16
    elif row['runs'] >= 50 and row['runs'] < 100:
        score = score + 8
    elif row['runs'] >= 30 and row['runs'] < 50:
        score = score + 4
    elif row['runs'] == 0:
        score = score - 2

    if row['balls'] >= 10:
        if row['sr'] > 170:
            score = score + 6
        elif row['sr'] > 150 and row['sr'] <= 170:
            score = score + 4
        elif row['sr'] > 130 and row['sr'] <= 150:
            score = score + 2
        elif row['sr'] > 60 and row['sr'] <= 70:
            score = score - 2
        elif row['sr'] > 50 and row['sr'] <= 60:
            score = score - 4
        elif row['sr'] <= 50:
            score = score - 6
        else:
            pass

    if row['Is_Captain'] == 1:
        score = score*2
        

    return score
````

````python
final_df.apply(dream11 ,axis= 1) # every row
````

````python
export_df = final_df.sort_values('score',ascending=False)[['ID','batter','score']]
````
