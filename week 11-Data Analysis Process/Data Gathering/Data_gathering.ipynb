## export to_html

````python
sixes =  df.query('batsman_runs== 6')
````

````python
# ipl ki har ball par kitna sixes lagta ha 
overs = [1,2,3,4,5,6]
batsman_sixes = sixes[sixes['ball'].isin(overs)]
````

````python
sixes_df = batsman_sixes.pivot_table(
    index='over',
    columns='ball',
    values='batsman_runs',
    aggfunc='count'
)
sixes_df
````
