### find the last match played by virat kohli in delhi 
Thinking process
- find the virat kohli in team1 players and team2 players
- find when city is Equal to delhi
- find the last match

#### my sol

```python
ipl.head()
ipl['Team1Players'].str.contains('V Kohli') 
ipl['Team2Players'].str.contains('V Kohli') 
mask = (ipl['Team1Players'].str.contains('V Kohli')) | (ipl['Team2Players'].str.contains('V Kohli'))
ipl['City'] =='Delhi'
ipl[mask | (ipl['City'] =='Delhi')]
```

```python
ipl['City'] =='Delhi'
ipl[(ipl['City'] =='Delhi') &  mask].sort_values(by='Date',ascending=False).head(1)
```

#### sir solution 
- we make a new col of 22 player add team1 player + team2 player
- ek function banya ga did_kohli_play 
- check city is delhi
- apply the drop using subset city,did kohli play

```python
# we does not want to make permanent changes in df because add col
# we make a copy
cricket_df = ipl.copy()
cricket_df['all_player'] = cricket_df['Team1Players'] + cricket_df['Team2Players']
cricket_df.head()
```

```python
def did_kohli_play(player_list):
    return 'V Kohli' in player_list
# apply our custom function ki 22 player ma kohli ha ki nahi 
cricket_df['did_kohli_play']= cricket_df['all_player'].apply(did_kohli_play)
cricket_df.head()
```

```python
# now apply the operation and when both conditions are true
cricket_df[(cricket_df['City'] =="Delhi") & (cricket_df['did_kohli_play'] == True)]
```

```python
mask = cricket_df[(cricket_df['City'] =="Delhi") & (cricket_df['did_kohli_play'] == True)]
mask.drop_duplicates(subset=['City','did_kohli_play'],keep='first')
# we finally fin the answer 
```

```python
points_df = pd.DataFrame(
    {
        '1st point':[(3,4),(-6,5),(0,0),(-10,1),(4,5)],
        '2nd point':[(-3,4),(0,0),(2,2),(10,10),(1,1)]
    }
)

points_df
```

```python
# so abhi tak hamna deka apply top to down har ek coloumn par 
# kam kar raha tha lakin ham chaya har ek row par bhi laga sakta ha 
# input ma row mil raha ha 
def euclidean(row):
    point_a = row['1st point']
    point_b = row['2nd point']
    expression = ((point_a[0] - point_b[0])**2 + (point_a[1] - point_b[1])**2)**0.5
    return expression
points_df.apply(euclidean, axis= 1)
```

```python
# now make a new coloumn 
points_df['distance'] = points_df.apply(euclidean, axis= 1)
points_df
```
