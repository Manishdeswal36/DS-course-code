### toss winner or match winner in percentage
hum dekna chata ha ki toss jitna sa match winner ma kuch hota ha ki nahi 

```python
# find the no of  toss winner and wining team and  are same
# hum dono coloumns ko compare kar dega 
ipl['TossWinner'] == ipl['WinningTeam']
# True toss jita match jita , False toss jita,matcha hara 
# use indexing concept 
ipl[ipl['TossWinner'] == ipl['WinningTeam']]
# find number of rows using shape 
ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0]
# divide the ratio
ratio = ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0]/ipl.shape[0]
ratio * 100
print(" {}% times toss wining and match wining team is same ".format(round(ratio * 100)))
```

### write a function that give you track record of 2 teams against each other
example yadi ma apko 2 teams du rcb vs csk tho rcb na kitna match jita or csk na kitna match jita 

```python
# thinking behind func
# input ma two teams  output ma how much match won between them

ipl.sample()
['Team1','Team2','WinningTeam']
# input ma team1 and team2
# fir un dono team ma matches kab hua 
# fir usma sa winning team first thi or second thi 
# fir shape sa count kar lao 
def win_matches(first_team,second_team):

    ipl['Team1'] == first_team
    ipl['Team2'] == second_team
    ans1 = ipl[(ipl['Team1'] == first_team) & (ipl['Team2'] == second_team) & (ipl['WinningTeam']== first_team)].shape[0]
    ans2 = ipl[(ipl['Team1'] == first_team) & (ipl['Team2'] == second_team) & (ipl['WinningTeam']== second_team)].shape[0]
    
    print('{} won =  {} matches'.format(first_team,ans1))
    print('{} won =  {} matches'.format(second_team,ans2))
    
win_matches(first_team = 'Royal Challengers Bangalore',second_team = 'Chennai Super Kings')
```
