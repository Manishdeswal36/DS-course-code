# Merge

## table preview


```python
pets = pd.DataFrame(
    data={
        'name': ['Mr. Snuggles', 'Honey Chew Chew', 'Professor', 'Chairman Meow', 'Neighbelline'],
        'type': ['cat', 'dog', 'dog', 'cat', 'horse']
    },
    index=[71, 42, 11, 98, 42]
)

visits = pd.DataFrame(
    data={
        'pet_id': [42, 31, 71, 42, 98, 42],
        'date': ['2019-03-15', '2019-03-15', '2019-04-05', '2019-04-06', '2019-04-12', '2019-04-12']
    }
)
display(pets)
display(visits)
```

```python
pd.merge( left=pets,
        right=visits,
        how='inner',
        left_index=True,
        right_on='pet_id'
        )
```

## inner join
always  do good names before join table it will make structed data
we reanme the indexes so the indexes are in sequence

```python
pets.index.rename('pet_id',inplace= True)
display(pets)
```

```python
visits.index.rename('visits_id',inplace= True)
display(visits)
```

```python
pd.merge(
    left= pets,
    right= visits,
    how= 'inner',
    on= 'pet_id'
)
#benefit we have a sequence of indexes
#now more readable
#inner intersection b/w two sets
```

## left join

```python
pd.merge(
    left= pets,
    right= visits,
    how= 'inner',
    on= 'pet_id'
)
#notice we have nan value beacuse pandas didn't find date value in right table
```

## Right join

```python
pd.merge(
    left=pets,
    right= visits,
    how = 'right',
    on = 'pet_id'
)
#right table + intersection(common)in left table
#nan in name and type
```

## Outer join 

```python
pd.merge(
    left=pets,
    right= visits,
    how = 'outer',
    on = 'pet_id'
)
#includes all keys in both table
#union of all sets
```

## Anti join
results includes only keys from the left table not the right table

### method one

```python
#results includes only from the one table not others (inverse of one and another)
#Which records from table A don't match any records from table B
outer =pd.merge(
    left= pets,
    right= visits,
    how= 'outer',
    on= 'pet_id',
    indicator= True,
    
)
display(outer)
```

```python
outer._merge == 'left_only'
#find left only with boolean
```

```python
a_not_in_b = outer.loc[outer._merge == 'left_only']
display(a_not_in_b)
#wow this is good you find it but sometimes it memory effiecient 
# to create a big table in intermediate step
```

### mehod 2
The basic idea is, for each pet_id in the pets table, we check if it exists in the visits table using the isin() method. Then we negate the result and use that to subset the pets table.

```python
pets.index.isin(visits.pet_id)
pets.index.isin(visits['pet_id'])
#both are same you use .  operator or [] 
#bollean array means -> kon kon shi value exists karti means(ha same dona ma comman ha )
```

```python
#we do neagtiation it because array display karta samha true values print hoti ha false nhi 
~(pets.index.isin(visits.pet_id))
```

```python
# we put indexes using loc 
pets.loc
print(pets.loc[~(pets.index.isin(visits.pet_id))])
```

```python
#if it have duplicate values so we use unique method 
print(pets.loc[~(pets.index.isin(visits.pet_id.unique()))])
```

# Agrreation
it is used for summmary statistics for datframe

```python
df6 = pd.DataFrame({
    'x': [3.1, 5.5, 9.2, 1.7, 1.2, 8.3, 2.6],
    'y': [1.4, np.nan, 5.0, 5.8, 9.0, np.nan, 9.2]
})
display(df6)
```

```python
df6.aggregate('sum')
df6.aggregate('mean')
```

```python
#don't call the function or don't use brackets ins
df6.aggregate(np.median)
#this is a series you make make dataframe by passing in list
```

```python
df6.aggregate([np.median])
```

```python
# you can also use your own custom function 
def myfunc(z):
    return z[1] 
```

```python
#when you use don't use bracket
df6.aggregate(myfunc)
```

```python
df6.agg(['mean','median'])
```

```python
df6.agg({
    'x' :['mean','sum'],
    'y' :['min', 'max']  
})
```

```python
#by default axis = 0 means column wise you also do rows wise axis = 1
df6.agg(['median'], axis=1)
```

# group by
you can use groupby to parition your dataframe or series and subsequentialy you can aggerate or tranform to it 
- parition(dividing a space into two or many parts)
- aggerate(do statistics & use functions o rows and columns as your choice)

```python
df = pd.DataFrame({
    'A': ['foo', 'bar', 'foo', 'bar', 'bar', 'foo', 'foo'],
    'B': [False, True, False, True, True, True, True],
    'C': [2.1, 1.9, 3.6, 4.0, 1.9, 7.8, 2.8],
    'D': [50, np.nan, 30, 90, 10, np.nan, 10]
}).convert_dtypes()  
#convert_dtypes is a handy way to make sure we're using pandas' modern String and Integer datatypes
# which allow for NaN values. Without it, A would be type 'object', and D would be type 'float64'.
display(df)
```

```python
df.groupby(by="A")
#df.groupby(by='A') tells pandas to partition the data based on unique values in column A.
```

```python
#like this 
        # |     |  A  |   B   |  C  |   D  |
        # |:---:|:---:|:-----:|:---:|:----:|
        # |   1 | bar | True  | 1.9 | <NA> |
        # |   3 | bar | True  |   4 | 90   |
        # |   4 | bar | True  | 1.9 | 10   |
        # |-----|-----|-------|-----|------|
        # |   0 | foo | False | 2.1 | 50   |
        # |   2 | foo | False | 3.6 | 30   |
        # |   5 | foo | True  | 7.8 | <NA> |
        # |   6 | foo | True  | 2.8 | 10   |
```

```python
group_ab = df.groupby(by=['A',"B"])
group_ab.groups
#it tells you that ('bar',True)
#at 1,3,4 index
```

```python
group_ab.ngroup()
#ngroup() tells us how many partions occurs
#this are indexes that divides in groups
#  row_index  = groups A
# [1,3,4]-> group 0
# [0,2] -> group 1
# [5,6] -> group 2
```

```python
wholes = df.C.apply(np.floor)
display(wholes)
```
