# bees and knees

```python
import numpy as np
import pandas as pd

bees = pd.Series([True, True, False, np.nan, True, False, True, np.nan])
knees = pd.Series([5,2,9,1,3,10,5,2], index = [7,0,2,6,3,5,1,4])

print(bees)
print(knees)
```

```python
#find nan using isnull
pd.isnull(bees)
#convert to numpy
pd.isnull(bees).to_numpy()
#use index postion to fit
knees.loc[pd.isnull(bees)]
```

```python
# do power of 2
res = knees.loc[pd.isnull(bees).to_numpy()] =knees.loc[pd.isnull(bees).to_numpy()] **2 
print(res)
```

Price gouging

```python
import numpy as np
import pandas as pd

generator = np.random.default_rng(123)
beef_prices = pd.Series(
    data = np.round(generator.uniform(low=3, high=5, size=10), 2),
    index = generator.choice(10, size=10, replace=False)
)
print(beef_prices)
```

solution

Goal = Determine which day had the biggest price increase from the prior day.       
1. we sort the days
2. shift the days by one .why?
3. beacause we measure the gap b/w prevous day and next day
4. gap is max find with idxmax method

```python
# sort indexes means days
prev_days = beef_prices.sort_index()
print(prev_days)
```

```python
#we can shift previous days
next_days = beef_prices.shift()
print(next_days)
```

```python
daily_changes = prev_days -next_days
daily_changes
```

```python
daily_changes.idxmax()
```
