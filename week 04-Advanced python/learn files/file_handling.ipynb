# Pickling
- `Pickling` is the process where a python object hierarchy is converted into a byte stream and
- `unpickling` is the inverse operation whereby a byte stream(from a binary file or vytes-like object) us converted back into an object hierachy

```python
# example apna vo bankclass aor fraction class banayi ha
# pickel --> your custom class to binrary ma convert kiya
# sent to another file
# unpickel --> now your object of bankclass or fraction behave same in another file
# jasa vo apni file ma kar raha tha 
```

```python
# koi class ka             Serlization                          De-serlization                           
#   object          -----> convert to another data type -----> fir us object ko                                                             
#                           ex dict, str ,list                  use karta
#



# ek object ko dursi file ma la jana ha tho usko data type ma convert karna hoga
# fir usa use kara 
```

```python
#  koi class ka                     kisi aor file ma            same function ka sath
#    object             -------->    Transfer karna  -------->  dursi file ma use karna
#  uski functionalities ha              chata hu                    chatu hu


#   ya kasa karo ?

# JSON
#           data_type_convert             data_type_convert

# object ---> dict,str -----> binary -----> dict,str -------> object

# but using pickle 
#
# object ---> binary ---> object 
# ( with same functinalites ek file sa dursi jaga )
```

## biggest power kya ha pickling ki

## convert your object to binary 

```python
class Human:

    def __init__(self,name, action):
        self.name = name
        self.action = action

    def show_info(self):
        print('my name is {} my action is {} '.format(self.name,self.action))

person = Human('manish','focus on persent')
```

```python
import pickle
with open('Human.pkl','wb') as file_object:
    pickle.dump(person,file_object)
```

```python
import pickle
with open('Human.pkl','rb') as file_object:
    sample_person = pickle.load(file_object)

sample_person.show_info() # hara wha
```

# pickle vs Json

```python
# both are use to serlization and De serlization technique (data transfer techniques)

# then when to use , what? 
# kab konsa use karna ha 

#           pickle                                      JSON
# jab apko object ka funcalites           jab apko object ko human ready able 
# (data or attribute) access karni ho     ya python ka ready made data types ki                                                                
#  dursi file ma then use it              funcalites use karni ho then use it                                                              
#                                                                                   
#  jab convert karta ha tho binary ma     isma apko python ka ready mad data typed                                                                     
#   hota ha                                ex  str ,dict ma convert karno hoga                                               
#      
#  pickel ma object apna class ka method   JSON ma object apna class ka method 
# ko use  kar sakta ha                      ko use nhi kar sakta ha funcalities over 
#                                         becuase you use python  in bulit data types
```

```python
# as a data scienctist or machine learning enginer 
# you use pickel a lot you make a  ml model on your machine
# then convert to binary and someone else machine use your object
```
