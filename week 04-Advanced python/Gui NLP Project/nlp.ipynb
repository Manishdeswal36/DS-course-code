```python
import torch
from tqdm import TqdmDeprecationWarning
from transformers import AutoModelForSequenceClassification, AutoTokenizer

# Load the tokenizer and model
model_name = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Sample text for semantic analysis
text = "I love using Hugging Face transformers! They make NLP so easy."

# Tokenize the input text
inputs = tokenizer(text, return_tensors="pt")

# Make sure the model is in evaluation mode
model.eval()

# Perform the prediction
with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits

# Get the predicted class
predicted_class = torch.argmax(logits, dim=1).item()

# Print the result
try:
    labels = ["negative", "positive"]  # Adjust according to your specific task or model
    print(f"Predicted sentiment: {labels[predicted_class]}")
except TqdmDeprecationWarning as t:
    print(t)
```
