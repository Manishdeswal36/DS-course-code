### Recurssion solution 

```python
# what things you do in Recurssion
# you do two things in Recurssion
# you find base case 
# 1.base case jiska value apko pahala sa pata rahega
# you doesnot need any calcultaion 
# 2. big promblem ko choti problem ma decompose karta ho 
```

```python
# decompose of promblem
# 3*4

# 3 + 3*3  # breakdown in one of decimal system

# 3 + 3 + 2*3

# 3 + 3 + 3 + 1*3

# 3 + 3 + 3 + 3   # then simply it add it   

# so generalise
# a*b
# b = 1,a*b,=>a
```

```python
def mul(a,b):
    if b == 0:   # power is 0 so equal to 1
        return 1
    elif b ==1 :  # power is 1 so equal to self 
        return a
    else:      # jitni times power ha unti bar add karo 
        return a + mul(a,b-1)

mul(4,5)
#working see in  python tutor
# function call 
# go top to down 
# down to top 
```

## in recurssion

```python
def factorial(number):
    # find first base cases
    if number ==1:
        return 1
    else:
        return number * factorial(number-1) # kyu
factorial(5)
```

# Palindrome

```python
# front sa read karo last sa read karo same aya ga
# example madam ,racecar, malyalam
# symmertic  from front and back 
# think of Base case any single string in palindrome (example a ,b f,g)
# if length of string ==1 palindrom 
# if not != 1 then you 
# check first and last character they match or not  
# if mathch then you reomve it 
# then you again check until 1 is reach
# example (m a d a m) 
# remove m from front and back 
# then check it (a d a )
# then reomve it (  d  )
# last sring left then palindrome
```

```python
# we use slicing and indexing extract concept 
def palindrome(text):
    if len(text) == 1:
        print('yes this string is palindrome')
    else:
        #first position equal to last position of text 
        if text[0] == text[-1]:
            palindrome(text[1:-1]) # call the function
        else:
            print('it is not palindrome')
        
palindrome('racecar')
palindrome('madam')
palindrome('python')
```
