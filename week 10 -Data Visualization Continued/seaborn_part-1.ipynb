# facet plot
- kisi ek category ka upar multiple plots bana sakta ho . 
- in general cateogry column ma jitni bhi catgoery ha utna graph draw karo side by side  aap row or column dono ma kar saktah ho
- facet plot work only in figure level function not work in axis level function
- meaning it work only in relplot not in scatter and line plot

```python
# example you draw a scatter plot
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex')
```

```python
# ya bhut acha how much bill and tip pay by males and females (by gender)
# lakin apka manager bolta ha 
# acha hota ki males alag and females alag scatter plot hota 
# tho hum acha sa dek pata 
# solution is facetplot => kisi ek category pat multiples plots bana sakta ho
```

```python
sns.relplot(
    data=tips,
    x='total_bill', y='tip', # both numerical col
    kind='scatter', # type of graph
    col='sex' # sex column ma jitni bhi catgoery ha utna graph draw karo
# side by side 
)
```

```python
sns.relplot(
    data=tips,
    x='total_bill', y='tip', # both numerical col
    kind='scatter', # type of graph
    hue='sex', # colour by sex(gender) cateogory
    col='sex' # sex column ma jitni bhi catgoery ha utna graph draw karo
)
```

```python
sns.relplot(
    data=tips,
    x='total_bill', y='tip', # both numerical col
    kind='scatter', # type of graph
    row='sex' # sex column ma jitni bhi catgoery ha utna graph draw karo
# upar and nicha  
)
```

```python
# in general cateogry column ma jitni bhi catgoery ha 
# utna graph draw karo side by side 
# aap row or column dono ma kar saktah ho
```

```python
# doing both rows and coloumn
sns.relplot(
    data=tips,
    x='total_bill', y='tip', # both numerical col
    kind='scatter', # type of graph
    row='sex' ,
    col='smoker'
)
```

```python
# doing both rows and coloumn
sns.relplot(
    data=tips,
    x='total_bill', y='tip', # both numerical col
    kind='scatter', # type of graph
    row='day' ,
    col='sex'
)
```

```python
# also do in line plot
sns.relplot(
    kind='line',
    data=four_country_df,
    x='year',
    y = 'lifeExp',
    col='country' # by each country 
)
```

### col wrap

```python

sns.relplot(
    kind='scatter',
    data= gap,
    x= gap['gdpPercap'],
    y= gap['lifeExp'],
    col='year',
    col_wrap=4
)
```
