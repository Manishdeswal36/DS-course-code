## is

```python
# it compare the memory adrress in the object and give true or false 
a = 5
b = a 

print("id of a ", id(a))
print("id of b ", id(b))
```

```python
var_1 = [10,20,30]
var_2 = var_1[:] # cloning 
print(var_2)
print("id of var_1 ", id(var_1))
print("id of var_2 ", id(var_2))
id(var_1) == id(var_2) # their memory address is not same (differnet )
# but 
var_1 == var_2 # thier value is same 
```
