Conditional probability and joint probability are closely related but fundamentally distinct concepts.          
Joint probability measures the likelihood of two events happening together, while conditional probability measures the likelihood of an event occurring given that another event has already occurred.               
They are connected, but conditional probability is the joint probability divided by the probability of the known event

```python
# find the conditoanl proability by normalise = columns
pd.crosstab(df['Outlook'], df['Play Tennis'],normalize='columns')
```

```python
pd.crosstab(df['Outlook'], df['Play Tennis'],normalize='columns').stack().to_dict()
```

Joint probability is the probability of two events occurring simultaneously.
Marginal probability is the probability of an event irrespective of the outcome of another variable.
Conditional probability is the probability of one event occurring in the presence of a second event.

```python
probability_dict = {}
for i in ['Outlook','Temperature',	'Humidity',	'Wind']:
    probability_dict.update(pd.crosstab(df[i], df['Play Tennis'],normalize='columns').stack().to_dict())

probability_dict
```

```python
pd.Series(probability_dict)
```

```python
pd.Series(probability_dict).unstack(level=0)
```
