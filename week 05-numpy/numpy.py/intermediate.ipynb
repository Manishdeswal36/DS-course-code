### what is broadcasting 

#### 1 dim 

```python
# probaly you see it but don't realise
# example when add a vector(1d array) to a scaler
import numpy as np
np.array([1,2,3]) + 0.5 
# in the essence numpy is expanding the sclaer than add it
# like this
# [1,2,3] + [0.5,0.5,0.5]
# element wise additon between the arrays
# under the hood numpy does not actually do it is horrible in efficient
# but in the essence what's happeing 
# 1 + 0.5 = 1.5
# 2 + 0.5 = 2.5
# 3 + 0.5 = 3.5
np.array([1,2,3]) + np.array([0.5,0.5,0.5])
```

#### 2 dim 

```python
foo = np.array([
    [1,1,1],
    [2,2,2],
])

boo = np.array(
    [1,0,-1]
)
# broadbcasting  (2d array ) 
# matrix and a vector  it convert vector to matrix 
#  before      -->      after 
# (2,3) + (1,3) --> (2,3) + (2,3)
# match the shape and do element wise operation
# in essence it do like this 
[1,1,1] + [1,0,-1]
[2,2,2] + [1,0,-1]
foo + boo 
```

### how can broadcasting work and when we use it 

#### broadcasting rules 
suppose we want to add two array A and B ...
1. Moving backwards from the last dimension of each array,
2. We check if their dimensions are "compatible". Dimensions are compatible if they are equal or either of them is 1
3. If all of A's dimensions are compatible with B's dimensions, or vice versa, they are compatible arrays

1. (m,n) +(1,n) = (m,n)
2. (1,n) +(m,n) = (m,n)
3. (1,n) +(1,m) = (m,n)

#### 1st example 

```python
np.random.seed(1234)
A = np.random.randint(low =1 ,high= 10, size=(3,4))
B = np.random.randint(low =1 ,high= 10, size=(3,1))

#                           A           B
# A : (3,4)            [[4 7 6 5]      [[7]
# B : (3,1)             [9 2 8 7]  +    [3]
#                       [9 1 6 1]]      [1]]

# numpy  compare it's last dimesnion A : 4 and B : 1 then it convert B 4
# now last 2nd element is 3 so they are compatiable 
# now left is whatever procedure we wanted to do  
#                           A           B
# A : (3,4)            [[4 7 6 5]      [[7,7, 7, 7]
# B : (3,4)             [9 2 8 7]  +    [3,3, 3, 3]
#                       [9 1 6 1]]      [1,1, 1, 1]]
print( A +B)
```

### before

# Boolean indexing
one of powerful feature in numpy is boolean indexing 
With boolean indexing, you can subset an array A using another array B of boolean values.

```python
import numpy as np
foo = np.array([
    [3, 9, 7],
    [2, 0, 3],
    [3, 3, 1]
])
foo
```

#### we want to replace every 3 wih zero 

```python
#foo ==3
#both are same to do subset
mask = foo ==3
print(mask)
# true means on that index position is 3 
# false means on that index position 3 is not persent 
```

```python
#foo[foo ==3]
print(foo[mask])
```

```python
# now setting thus variable to 0 
foo[mask] = 0
print(foo)
```

### pick specific elements 

```python
rows_1_and_3 = np.array([True,False,True]) # like foo([0,2])
columns_1_and_2 =  np.array([False,True,True]) # like foo(:,[1,2])
```

```python
foo[rows_1_and_3,:] #like foo[[0,2],:] object[[row_index,col_index]]
```

```python
foo[:,columns_1_and_2 ] #like foo[:,[1,2]]
```

### how numpy treat boolean indexing 

```python
# you will be suprised by the result one -d array
# actually numpy treat boolean indices like integers .
# where the integer used are the true indices of element 
# numpy treat index 
# rows_1_and_3= [true,false, true] just like he treat index [0,2]
# columns_1_and_2= [false,true,true] just like he teate index [1,2]
foo[rows_1_and_3,columns_1_and_2] #foo([0,2],[1,2])
# select element
#(0,1) and (2,2)
```

# Logical operators

```python
b1 = np.array([False, False, True, True])
b2 = np.array([False, True, False, True])

b1 & b2  # [False, False, False,  True], and, intersection
b1 | b2  # [False,  True,  True,  True], or, union
b1 ^ b2  # [False,  True,  True, False], xor
~np.array([False, True]) # array([ True, False]) inverse
```

### sample rows from a 5*2 matrix

```python
foo[rand_rows]
# an array of random rows selected from foo 
# this is with replacement if you want to do without replacement 
# use choice function 
```
