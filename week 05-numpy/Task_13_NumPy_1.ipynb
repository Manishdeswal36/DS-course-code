You are given a space separated list of numbers. Your task is to print a reversed NumPy array with the element type float.



**Input Format:**

A single line of input containing space separated numbers.

**Output Format:**

Print the reverse NumPy array with type float.

**Example 1:**


Create a Python function to calculate the Softmax of the given numpy 1D array. The function only accepts the numpy 1D array, otherwise raise error.

$$\Large \sigma(\vec z)_i = \frac{e^{z_i}}{\sum_{j=i}^{K}{e^{z_j}}}$$

https://en.wikipedia.org/wiki/Softmax_function



**Example 1:**

Input:
```bash
[86.03331084 37.7285648  48.64908087 87.16563062 38.40852563 37.20006318]
```

Output:
```bash

       1.39571531e-10, 9.99896577e-01]
```

````python
class OneDimensionError(Exception):
    def message(self):
        print('please enter only one dimension array not higher dimension')

    
def softmax_function(given_array: np.ndarray) -> np.ndarray:
    array_var = np.array(given_array)
    if array_var.ndim == 1:
        expression =  np.exp(array_var)/np.sum(np.exp(array_var))
        print(expression)
    else:
        raise OneDimensionError

input_array = np.array([86.03331084 ,37.7285648  ,48.64908087 ,87.16563062, 38.40852563, 37.20006318])
softmax_function(input_array)
````

###`Q-19:` Write a function which will accept 2 arguments.
First: A 1D numpy array arr

Second: An integer n {Please make sure n<=len(arr)}

Output: The output should be the nth largest item out of the array
```
# Example1 : arr=(12,34,40,7,1,0) and n=3, the output should be 12
# Example2 : arr=(12,34,40,7,1,0) and n=1, the output should be 40
```

````python
# take two arugements 1d  numpy array and int -> uski length uska barabar ya kam honi chaya
# jis number par maximum aya vo extract (iska solution sort karo aor indexing nikalao)
def find_max(numpy_array:np.ndarray ,n:int) -> int:
    if  n<=len(numpy_array):
        sorted_array = np.sort(numpy_array)[::-1]
        print(sorted_array[n-1]) # it start from 0 and n is length so we do n-1 for index 
    else: 
        print('please enter number in the array range ')
find_max(numpy_array=(12,34,40,7,1,0),n= 3)
````

````python
find_max(numpy_array=(12,34,40,7,1,0) , n=1)
````
