###`Q-1:` Find the nearest element in the array to a given integer.
```
Ex:-
a=23 and array - [10 17 24 31 38 45 52 59].
Nearest element is 24
```
Hint: Read about this function `argmin()`

````python
import numpy as np 
def find_nearest(array: np.ndarray, value: int) -> int:
    idx = (np.abs(array - value)).argmin() # find index of the nearest value
    return array[idx]
input_array = np.array([10 ,17, 24, 31 ,38 ,45 ,52, 59])
find_nearest(input_array, 23) # 2
````

````python
array = np.array([10 ,17, 24, 31 ,38 ,45 ,52, 59])
value = 23
print(array)
print((array - value))
print(np.abs(array - value))
idx = np.abs(array - value).argmin()
print(idx)
print(array[idx])
````

###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])
np.unique(arr,return_counts=True,)
````

````python
# use the assignment concept 
arr[np.isnan(arr)] = 2
arr
````

````python
import numpy as np
arr = np.array([[1, 2, np.nan], 
                [4, 2, 6], 
                [np.nan, np.nan, 5]])

# Flatten the array and remove NaN values
filtered_arr = arr[~np.isnan(arr)]  
# Find the most frequent element
unique_elements, counts = np.unique(filtered_arr, return_counts=True)
most_common_element = unique_elements[np.argmax(counts)]  
# Replace NaN values with the most common element
arr[np.isnan(arr)] = most_common_element
print(arr)
````

### `Q-9:` Write a program implement Boradcasting Rule to check if two array can be added or not.
Given tuples of shapes.
```
shape of a- (3,2,2)
shape of b- (2,2)

check_broadcast(a, b) -> return Boolean (True if can broadcasted, False other wise.)
```

````python
def check_broadcasting(arr1: np.ndarray, arr2: np.ndarray) -> bool:
    try:
        arr1 + arr2
    except ValueError('array are not Broadcastable'):
        return False
    return True
a = np.arange(12).reshape(3,2,2)
b = np.arange(4).reshape(2,2)
check_broadcasting(a,b)
````
