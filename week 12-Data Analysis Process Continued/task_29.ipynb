```python
import plotly.express as px

fig = px.histogram(df, x='age', nbins=10, opacity=0.7,histnorm='density')
fig.update_traces(marker_line_color='black', marker_line_width=1.2)
fig.update_layout(bargap=0.05, title='Improved Plotly Histogram')
fig.show()
```
