## fill missing values

```python
df.isnull().sum()
```

```python
df.corr(numeric_only=True)['rating']
```

```python
# ratings column depend on many columns 
# kyuki hardware ko rate kareka hi ham rating nikal raha ho 
# rating ma missing values ha
# lakin apko usa sursa column sa hisab sa match comparsion relation nikal kar 
# fill karn ha
# 25 variable ha tho 25 dimension ma lakar usko fill karna ha 
# isko bolta ha multivarite immuniation of using knn immunicatot 
```

```python
our_df = df.select_dtypes(include=['int64','float64']).drop(columns='price')
our_df
```

```python
# vo sabhi column ma missing values fill kar dega
from sklearn.impute import KNNImputer
imputer = KNNImputer(n_neighbors=5)
imputer.fit_transform(our_df)
```

```python
# before 
our_df.values
```

```python
pd.DataFrame(imputer.fit_transform(our_df)).isnull().sum()
```

```python
## assign them 
our_df.values = imputer.fit_transform(our_df)
```

```python
pd.DataFrame(data=imputer.fit_transform(our_df),columns=our_df.columns)
```

```python
# yes our new data frame create
```

```python
pd.DataFrame(data=imputer.fit_transform(our_df),columns=our_df.columns).isnull().sum()
```

```python
# asign it 
our_df = pd.DataFrame(data=imputer.fit_transform(our_df),columns=our_df.columns)
```

```python
our_df.isnull().sum()
```
