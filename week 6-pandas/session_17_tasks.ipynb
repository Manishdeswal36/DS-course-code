###`Q-8:` Team1 vs Team2 Dashbord
Create a function which will take two string(name of two teams) as input. Show win Loss record between them and player getting most player of the match award in matches between these two teams.
```
team1_vs_team2('Kolkata Knight Riders','Chennai Super Kings')
```

````python
ipl.sample()
['Team1','Team2','WinningTeam']
# input ma team1 and team2
# fir un dono team ma matches kab hua 
# fir usma sa winning team first thi or second thi 
# fir shape sa count kar lao 
def team1_vs_team2(first_team,second_team):

    ipl['Team1'] == first_team
    ipl['Team2'] == second_team
    ans1 = ipl[(ipl['Team1'] == first_team) & (ipl['Team2'] == second_team) & (ipl['WinningTeam']== first_team)].shape[0]
    ans2 = ipl[(ipl['Team1'] == first_team) & (ipl['Team2'] == second_team) & (ipl['WinningTeam']== second_team)].shape[0]
    
    print('{} won =  {} matches'.format(first_team,ans1))
    print('{} won =  {} matches'.format(second_team,ans2))
    
team1_vs_team2(first_team = 'Kolkata Knight Riders',second_team = 'Chennai Super Kings')
````

### `Q-9:` Find out the top 7 cities where the matches of Kolkata Knight Riders are played frequently and plot the result as bar chart.

*`.plot(kind = "bar")` can help you to plot the bar chart. Also you can learn more about this method from [here](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.html)*



````python
# agar team 1 and team 2 ma kabhi bhi apko kolkt knigh rider mila usa print karo 
ipl['Team1'] == 'Kolkata Knight Riders'
ipl['Team2'] == 'Kolkata Knight Riders'
# kabhi bhi ya dono event ho tho print he meaning use or operato 
matches = ipl[(ipl['Team1'] == 'Kolkata Knight Riders') | (ipl['Team2'] == 'Kolkata Knight Riders')]
matches['City'].value_counts().head(7).plot(kind='bar')
````

### `Q-10:` Find out the average margin for the team Mumbai Indians for only the session 2011.

````python
ipl['Team1'] == "Mumbai Indians"
ipl['Team2'] == "Mumbai Indians"
# kabhi bhi ya dono event ka bitwise or operator 
matches = ipl[(ipl['Team1'] == "Mumbai Indians") | (ipl['Team2'] == "Mumbai Indians")]
# season 2011 hona chaya 
# fir usma sa Margin col nikalo
# uska mean lo using mean function 
matches[matches['Season'] == '2011']['Margin'].mean()
````
