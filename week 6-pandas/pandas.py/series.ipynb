### logical operation

```python
ages = pd.Series(
    data = [42, 43, 14, 18, 1],
    index = ['peter', 'lois', 'chris', 'meg', 'stewie']
)
print(ages)
```

```python
genders = pd.Series(
    data = ['female', 'female', 'male', 'male', 'male'],
    index = ['lois', 'meg', 'chris', 'peter', 'stewie'],
    dtype = 'string'
)
print(genders)
```

```python
# identifying males younger than 18 like this.
(ages >18) & (genders == 'male')
ages.loc[(ages <18) & (genders == 'male')]
```

# Series vectazation 

```python
vector= pd.Series(np.random.normal(size = (10**6)))
vector
```

```python
def average(x):
    avg = 0.0
    for i in range(x.size):
        avg += x.iloc[i]/x.size
        
    return avg
```

```python
average(vector)
```

```python
vector.mean()
# see numpy is 10 times faster 
```

# Apply func

```python
bamboo= pd.Series([3,9,2,2,8,7])
bamboo
```

```python
def my_func(x):
    if x %2 ==0:
        return x-1  
    else :
        return x+3
```

```python
bamboo.apply(my_func)
```

# where

```python
np.random.seed(10)
x2 = pd.Series(np.random.randint(low=1,high=7,size=(10**7)))
x2
```

```python
x2.apply(my_func)
```

```python
result = pd.Series(np.where(x2%2==0, x2-1, x2+3))
result
#numpy is fast it gives you boost
```
