# Sting operations
it same like python string operations       
if you want to apply str method to all series use str. and then str method or your method


```python
#it is useful you have caterogrical or text data
import numpy as np
import pandas as pd
cajan = pd.Series([
    'gumbo',
    'crawfish boil',
    'Mardi Grass',
    'pirogue', 
    pd.NA, 
    'Zatarains'
],dtype='string')
display(cajan)
```

access string methods

```python
cajan.str
# it is now string object 
```

lowercase all strings 

```python
cajan.str.lower()
```

uppercase all strings

```python
cajan.str.upper()
```

find the length of characters

```python
cajan.str.len()
```

```python
cajan.str.split()
# it make each str to list
```

get the first string after splitting by 

```python
cajan.str.split().str.get(0)
```

replace str with another

```python
cajan.str.replace(pat='',repl='-',)
```

```python
cajan.str.replace(pat=' ',repl='-',)
#notice space is more then previous 
```

concentate using cat

```python
cajan.str.cat(sep='_')
```

check startswith

```python
cajan.str.startswith('p')
```

```python
cajan.str.endswith('s')
```

contains

```python
cajan.str.contains('bo')
```

use regex with extract

```python
cajan.str.extract(r'(\b[A-Z][a-z]+\b)')
```
