## A user can find the distance b/w two points 

````python
class Point:

    def __init__(self,x, y) :
        self.x_cod = x
        self.y_cod = y 

    def __str__(self):
        return  "<{},{}>".format(self.x_cod,self.y_cod)
    
    def eucilden_distance(self, other):
        distance = ((self.y_cod -other.y_cod)**2 + (self.x_cod -other.x_cod)**2)**0.5
        return distance
````

## A user can find shortest distance b/w  a 2d point and a line 

````python
# d = Ax+ By +c/ (A**2 + B**2)**0.5
````

````python
class Point:

    def __init__(self,x, y) :
        self.x_cod = x
        self.y_cod = y 

    def __str__(self):
        return  "<{},{}>".format(self.x_cod,self.y_cod)
    
    def eucilden_distance(self, other):
        return  ((self.y_cod -other.y_cod)**2 + (self.x_cod -other.x_cod)**2)**0.5
        
    def distance_from_origin(self):
        return ((self.y_cod -0)**2 + (self.x_cod - 0)**2)**0.5
# make another class
class Line:

    def __init__(self,A,B,C): # coefficint of line 
        self.A = A # they are different set of variable but we give name same
        self.B = B  # self.A = A are different variables 
        self.C = C
    
    def __str__(self): # our line looks like 3x + 2y + 4 =0
        return "{}x + {}y + {} = 0".format(self.A,self.B,self.C)
    
    # we are creating func that check
    def point_lie_line(line, point): # type: ignore 
        # humna objects ka  name self ya other na dekar line and point diya 
        if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
            return 'lies on line '
        else:
            return 'does not line on line'
        
    def shortest_distance(line , point):# type: ignore 
        return abs(line.A*point.x_cod + line.B * point.y_cod +line.C)/(line.A**2+line.B**2)**0.5
line_1 = Line(2,3,4)
point1 = Point(2,3)
line_1.shortest_distance(point1)
````
