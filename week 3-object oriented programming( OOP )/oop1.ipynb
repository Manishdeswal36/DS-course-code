

# Understanding `self` in Python Classes

## Observation
1. In all methods within a class, `self` is the default parameter.
2. Common methods include:
   - `__init__`
   - `menu`
   - `create_pin`
   - `check_pin`
   - `balance`
   - `withdraw`
3. `self` is used to reference the instance of the class itself.

## Key Points
1. **Constructor Method (`__init__`)**:
   - In the constructor method, all attributes are prefixed with `self`.
   - This is how attributes are initialized and associated with the object.
2. **Method Calls**:
   - Whenever you call a method on an object, `self` is implicitly passed as the first argument.
   - Example: `object.method()` is translated to `Class.method(object)` internally.

## Concept of `self`
1. **What is `self`?**
   - `self` represents the instance of the class.
   - It is similar to prefixes like "Shri" in names, giving a respectful identifier to the instance.

2. **Golden Rule of OOP**:
   - A class defines a set of rules (attributes and methods).
   - Attributes: e.g., `pin`, `balance`.
   - Methods: e.g., `menu`, `create_pin`, `withdraw`.

3. **Accessing Class Contents**:
   - The class's attributes and methods can only be accessed by the object of the class.
   - Example: Creating an object of `ATM` class and using the dot notation to access attributes and methods.

## Communication Between Methods
1. **Method Interaction**:
   - Methods within a class can interact with each other through `self`.
   - `self` acts as a medium for communication.
   - Example: `check_pin` does not directly call `withdraw`, but can communicate via `self`.

2. **Constructor and Method Calls**:
   - In the constructor, you cannot directly call another method without using `self`.
   - Example: `self.menu()` within `__init__`.

## Practical Demonstration
1. **ID Function**:
   - Using `id()` function, `self` and the object of the class have the same memory address.
   - This proves that `self` is indeed the current instance of the class.

2. **Experiment**:
   - Removing `self` from a method definition results in an error.
   - Example: If `self` is removed from `create_pin`, calling it results in `TypeError`.

## Deep Conceptual Understanding
1. **`self` is the current object**:
   - Represents the instance you are working with.
   - Ensures methods can interact with the instance's attributes and other methods.

2. **Naming Convention**:
   - The name `self` is a convention; it can be replaced with any other name, but `self` is widely used and understood.
   - Similar to `this` in Java.

3. **Intra-communication**:
   - Inside a class, variables and methods often need to communicate.
   - `self` is the means by which this intra-communication happens.

## Conclusion
- **`self`**: Ensures that methods and attributes within a class instance can interact.
- **Communication**: Methods use `self` to call other methods and access attributes.
- **Naming**: While `self` is a convention, it can technically be any name, but using `self` maintains clarity and consistency in your code.

````python
class Fraction:
    
    def __init__(self,x, y):
        self.num = x
        self.den = y 

    def __str__(self):
        return "{}/{}".format(self.num,self.den)
    
    def __add__(self, other): # lcm liya ha cross mulitply 
        new_num = self.num *other.den + self.den * other.num
        new_den = self.den * other.den
        return "{}/{}".format(new_num,new_den)
    
    def __sub__(self, other): # lcm liya ha cross mulitply 
        new_num = self.num *other.den -  self.den * other.num
        new_den = self.den * other.den
        return "{}/{}".format(new_num,new_den)
    
    def __mul__(self, other):
        new_num = self.num *other.num
        new_den = self.den * other.den
        return "{}/{}".format(new_num,new_den)
    
    def __truediv__(self, other):
        new_num = self.num *other.den
        new_den = self.den * other.num
        return "{}/{}".format(new_num,new_den)
    
    def convert_to_decimal(self):
        return  self.num/self.den
    
    def convert_to_percentage(self):
        return (self.num/self.den) *100
    

fr1 = Fraction(3,4,)
fr2 = Fraction(1,2)
````
