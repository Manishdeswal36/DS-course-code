````python
# now calcuate the m and b 
# now first calcuate slope
# check it ki mera data tik aa raha ha ki nahi
class MeraLR:
    # intilaise the constructor
    def __init__(self):
        self.m = None # slope
        self.b = None # y-intercept
        
    # create methods
    def fit(self, X_train,y_train):
        # calcuate slope
        num = 0
        den = 0
        for i in range(X_train.shape[0]):
            num = num + (X_train[i] - X_train.mean() * y_train[i] - y_train.mean())
            den = den + (X_train[i] - X_train.mean())**2
            
        self.m = num/den
        # calcuate y intercept
        self.b = y_train.mean() - self.m * X_train.mean()
        print('slope m =', self.m)
        print('y-intercept b = ',self.b)
        
        
    def predict(self,X_test):
        pass
    
our_model = MeraLR()
our_model.fit(X_train,y_train)
````
