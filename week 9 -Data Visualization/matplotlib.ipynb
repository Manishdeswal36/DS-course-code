```python
plt.hist(array , bins=[10,20,30,40,50,60,70],log=True)
plt.show()
```

## simple data 

```python
data = [23,45,100,20,69]

plt.pie(data)
plt.show()
```

## labels

```python
data = [23,45,100,20,69]
marks = ['eng','hindi','maths','sst','science']

plt.pie(data,labels=marks)
plt.show()
```

```python
df = pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'] ,labels=df['batsman'])

plt.show()
```

## percentage

```python
plt.pie(df['batsman_runs'] ,labels=df['batsman'],autopct='%0.1f%%')

plt.show()
```

```python
# This colour are generated automatic
# you can customise also 
```

## colours 

```python
plt.pie(df['batsman_runs'] ,labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','yellow','pink','cyan','brown'])

plt.show()
```

## extract particular pie 

```python
# example hum ab deviller ko dekaga
```

```python
plt.pie(df['batsman_runs'] ,labels=df['batsman'],autopct='%0.1f%%',explode=[0.2,0.01,0,0,0,0])

plt.show()
```

```python
plt.pie(df['batsman_runs'] ,labels=df['batsman'],autopct='%0.1f%%',explode=[0,0,0,1,0,0.03])

plt.show()
```

## shadow

```python
plt.pie(df['batsman_runs'] ,labels=df['batsman'],autopct='%0.1f%%',explode=[0.2,0.01,0,0,0,0],shadow= True)

plt.show()
```
