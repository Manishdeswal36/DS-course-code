````python
# python is so flexible ki vo apko bhi chance da raha ha apni
# custom Exception bana ka liya 
````

````python
# custom excpetion ban na ha tho apko inherit karna hoga Excpetion class sa
class MyException(Exception):
    def __init__(self,message):
        print(message)

class Bank:

    def __init__(self, balance):
        self.balance = balance

    def withrow(self,amount):
        if amount < 0:
            raise MyException('amount can not be -ve') #Exception ki jaga MyException
        if amount > self.balance:
            raise MyException('paisa nhi ha tera paas ') #Exception ki jaga MyException
        self.balance = self.balance - amount 

obj = Bank(10000)
try:
    obj.withrow(-5000)
except MyException as e:  # -ve ata hi except trigger hua 
    pass # apna uppar code lik raka ha MyExcpetion classe 
else:
    print(obj.balance)
````

Working of code 

````python
# -ve ata hi excpet trigger hua 
# except na MyException ko trigger kiya
# flow of program
# withraw call hua 
# fir condition check hui if amount > 0       -ve wali
# fir MyException class ma kya uska constructure activate hua 
# usna vo msg print kiya jo raise ka andr ha 
````

````python
# custom excpetion ban na ha tho apko inherit karna hoga Excpetion class sa
class MyException(Exception):
    def __init__(self,message):
        print(message)

class Bank:

    def __init__(self, balance):
        self.balance = balance

    def withrow(self,amount):
        if amount < 0:
            raise MyException('amount can not be -ve') 
        if amount > self.balance:
            raise MyException('paisa nhi ha tera paas ')
        self.balance = self.balance - amount 

obj = Bank(10000)
try:
    obj.withrow(500000) # balance sa jada amount deta hi MyException trigger 
except MyException as e: # except na MyException ko trigger kiya 
    pass 
else:
    print(obj.balance)
````

````python
class SecurityError(Exception):
    def __init__(self,message):
        print(message)

    def logout(self):
        print('logout kar da bhai sara account sa ')
class Google:
    
    def __init__(self , name, gmail,password,device):
        self.name = name
        self.gmail = gmail
        self.password = password
        self.device = device 

    def login(self, gmail_login, password_login, device_login):
        if device_login != self.device:
            raise SecurityError('tera tho lag gaya ')
        if gmail_login == self.gmail and password_login == self.password:
            print('login hua aor ')
        else:
            print('login erro tik sa detail bhar sa baba ')

obj = Google('Manish', 'manishdeswal@gmail.com', '1234', 'android')
try:
    obj.login('manishdeswal@gmail.com', '1234', 'windows')
except SecurityError as e:
    e.logout()
else:
    print(  'aa gya in the DS world ')
finally:
    print('database connection closed')
````

### login with same device 

````python
class SecurityError(Exception):
    def __init__(self,message): # error msg jo hum send 
        print(message) # SecurityError('tera tho lag gaya ') ya wala 

    def logout(self):
        print('logout kar da bhai sara account sa ')
class Google:
    # constructor basic info
    def __init__(self , name, gmail,password,device):
        self.name = name
        self.gmail = gmail
        self.password = password
        self.device = device 

    def login(self, gmail_login, password_login, device_login):
        if device_login != self.device: # yadi device_login login ka equal nahi ha tho 
            raise SecurityError('tera tho lag gaya ') # security error throw karo 
        if gmail_login == self.gmail and password_login == self.password:#varna check karo gmail and password
            print('login hua ha  ')
        else:
            print('login error tik sa detail bhar sa baba ')

obj = Google('Manish', 'manishdeswal@gmail.com', '1234', 'android')
try:
    obj.login('manishdeswal@gmail.com', '1234', 'android')
except SecurityError as e:  # catch error 
    e.logout()   # logout function exceute kar tho 
else:
    print('you are successful login {}'.format(obj.name))
finally:
    print('database connection closed')
````
