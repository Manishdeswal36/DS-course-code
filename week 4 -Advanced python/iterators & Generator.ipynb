# Making our own for loop

### create function

````python
# function ka andr iterator object bano ga
# fir ek infinte loop chalo ga 
# and try except ka andr
# jab tak loop chalaga jab tak item katam nhi hota
# agar item katam ho gaya then raise stopiteration Error 
````

````python
def mera_khudka_for_loop(iterator):

    iterator = iter(iterator)

    while True:
        try:
            print(next(iterator))
        except StopIteration:
            break
````

````python
test_on_list = [1,2,3]
test_on_tuple = (10,20,30)
test_on_range = range(1,10)
test_on_set = {4,5,6}
test_on_dict = {0:10 , 1:20}

# sata data type kam kar raha ha *
mera_khudka_for_loop(test_on_list)
# wow it work 
````

# Generatores
python generatores are a simply way to create generatores

````python
#iterable
class mera_range:

    def __init__(self,start,end):
        self.start = start
        self.end = end
        
    def __iter__(self):
        return mera_range_iterator(self)
    
# iterator
class mera_range_iterator:
    
    def __init__(self,iterable_object):
        self.iterable = iterable_object

    def __iter__(self):
        return self

    def __next__(self):
        if self.iterable.start >= self.iterable.end:
            raise StopIteration
        
        current = self.iterable.start
        self.iterable.start += 1
        return current
    
# deko class mera range ma
#   constructer
#   __iter__ method

# fir class mera_range_iterator
#   construnctor 
#   __iter__ method
#  ___next__ method*
# iteratar  banano pada  lakin ya old school tarika ha 
# ....
# tho python ka creators na isi cheez ko karna ka 
# tarika nikala lakin by simple syntax
# we create iterator but by generator concept 
````

### chaining Generator 

````python
# 2 generator ya 3 generator ko sath ma use kar sakta ho
# example apko fibonnaci sequence number lena ha aor uska square
# karna ha  fir unka sum 
````

````python
def fibonacci_number(number):
    # initialse the condition
    first_num,second_num = 0,1
    # loop through the number
    for _ in range(number):
        # use swaping  concept 
        first_num,second_num = second_num,first_num+second_num
        yield first_num
        
def square(number):
    for i in number:
        yield i**2

print(sum(square(fibonacci_number(10))))
````
