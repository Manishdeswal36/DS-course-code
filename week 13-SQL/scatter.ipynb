```python
import pandas as pd 
import plotly.express as px
gapminder_df = px.data.gapminder(year=2007)
gapminder_df
```

```python
px.scatter(
    data_frame=gapminder_df,
    x='gdpPercap',
    y='lifeExp',
)
```

```python
px.scatter(
    data_frame=gapminder_df, # our dataframe
    x='gdpPercap', #  x varible 
    y='lifeExp',  # y variable
    title= 'Health vs Wealth',
    labels={
        'gdpPercap': 'Wealth indicator', # old name : new name
        'lifeExp' : 'Health indiccator',
        }
)
```

```python
px.scatter(
    data_frame=gapminder_df, # our dataframe
    x='gdpPercap', # numerical variable 1
    y='lifeExp',  # numerical variable 2 
    title= 'Health vs Wealth',
    color='continent' # categorical varible 
)
```

```python
px.scatter(
    data_frame=gapminder_df, # our dataframe
    x='gdpPercap', # numerical variable 1
    y='lifeExp',  # numerical variable 2 
    title= 'Health vs Wealth',
    color='continent', # categorical varible 
    size= 'pop' # numerical variable 
)
```

```python
px.scatter(
    data_frame=gapminder_df, # our dataframe
    x='gdpPercap', # numerical variable 1
    y='lifeExp',  # numerical variable 2 
    title= 'Health vs Wealth',
    color='continent', # categorical varible 
    size= 'pop' ,# numerical variable 
    size_max=60
)
```
