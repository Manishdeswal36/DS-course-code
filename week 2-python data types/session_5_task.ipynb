###`Q3`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

````python
t1 = (1, 2, 3, 0)
t2 = (0, 1, 2, 3)
zip(t1, t2)
print(tuple(zip(t1, t2)))
# loop through each item check the item is same or different 

for (first_item, second_item) in zip(t1, t2):
    if first_item == second_item:
        continue
    print('not same')
    break
else:
    print('same')
````

### `Q3:` Write a program to Check if a given string is binary string of or not.

A string is said to be binary if it's consists of only two unique characters.

Take string input from user.

```
Input: str = "01010101010"
Output: Yes

Input: str = "1222211"
Output: Yes

Input: str = "Campusx"
Output: No
```

````python
def check_binary(given_string):
    #check given_string ko set ma convert karo  
    # set ma unique value hi aata ha then count the len 
        # if len of set is 2 binary otherwise not 
        convert_set = set(given_string) 
        if len(convert_set) == 2:  
            print("Yes it is  binary")
        else:
            print('not binary')

check_binary("01010101010")
````
