```python
# asa function jo input(argumnt) ma function recevice kara  
#            ya 
# asa function jo ouptut(return ma) function  sent kara 
# both situations ma koi bhi ek true this is higher order functio
```

make a function to calcuate square in a list

```python
def square(num):
    return num**2

# higher order func 
# input ma func and list la ga 
def transform(func, List):
    output = []          # list compreshenion logic 
    for i in List:
        output.append(func(i))
    
    print(output) 

try_list = [1,2,3,4,5]
transform(square, try_list)
```

```python
# ab muja cube karna ho tho function bano
def cube(x):
    return x**3
transform(cube, try_list)
```

but use lambda

```python
transform(lambda x:x**3, try_list) # wow so that'why lambd function use in higher order func
# one line ma logic express kar sak tha ho
# and func ka behavior change kar sakta ho
```
