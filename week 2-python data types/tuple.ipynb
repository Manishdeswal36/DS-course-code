speed test

```python
# computaion who can perform  fast compuation(operation)
# in 1 million  numbers/iterm
# how can we measure who is fast 
# the answer is time 
```

```python
# logic -->current time - past time
# loop start hua and loop ruka 
#kitna taj loop thoda
# stop time - start time 
```

```python
# concept  time classs ma andr time() method 
import time
time.time() #--> it actually measure current time 

start = time.time()
# .... after process happen 
stop =   time.time()- start
#change = final -  intial 
```

```python
import time
List = list(range(0, int(1e6))) #range(0, 1000000) 10*6
Tuple =tuple(range(0,int(1e6)))  #range(0, 1000000) 10*6

#loop speed measure
start = time.time()
for iter_var in List: 
    iter_var* 17 # any operation you want
print("list time" , time.time() - start)


#tuple speed measure
start = time.time()
for iter_var in Tuple:
    iter_var*17
print("tuple time" , time.time()  - start)

#tuple take less time 
```
