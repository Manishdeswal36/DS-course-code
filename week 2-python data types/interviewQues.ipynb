shallow copy

```python

a = [1,2,3,[4,5]]
#b = a.copy()
b = a[:]

print("id of a: ",id(a))
print("id of b: ",id(b))

b[-1][0] = 400
print(b)
print(a)
```

deep copy 

```python
import copy
a = [1,2,3,[4,5]]
b = copy.deepcopy(a)

print("id of a: ",id(a))
print("id of b: ",id(b))

b[-1][0] = 400
print(b)
print(a)
```
