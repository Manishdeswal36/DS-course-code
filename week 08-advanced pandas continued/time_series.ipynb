# Time series

A time series is a data set that tracks a sample over time. In particular, a time series allows one to see what factors influence certain variables from period to period. Time series analysis can be useful to see how a given asset, security, or economic variable changes over time.

Examples

- Financial Data (Company stocks,performance,sales)
- Natural Data (Rainfall measurement,temperature, any sensor Quanity)
- Event Data (Covid )
- Medical Data (Heart rate monitoring,BP(blood pressure) measurement,brain waves monitoring)

Types of Operations done on Time Series

- Time Series Analysis
- Time Series Forecasting

# Resampling

Resampling involves changing the frequency of your time series observations.

Two types of resampling are:

**Upsampling**: Where you increase the frequency of the samples, such as from minutes to seconds.<br>
**Downsampling**: Where you decrease the frequency of the samples, such as from days to months.

```python
# yadi aap apna time series data ka frequency change kar raho ho usa bola jata ha resampling
# tho apka dataset ka frequency kiya ha
# youtube wala ka daily
# google wala ka business 
```

```python
# Imp Question: why we need resampling
# agar apka data ek particular frequency par ha tho hum
# downsmapling ya upsampling kyu karni chaya.
# generally aap downsampling karta ho upsampling rare cases ma karta ho
```

### Downsampling
move from higher timeframe to lower timeframe apka dataset ka frequency

example apka data ka frequency `minute` ma ha 
apna usko `second` ma convert kar diya using asfrequency  

```python
google_df['Volume'].plot()
```

```python
## deko ya volume ka data thoda noisy ha hum trend identify nahi kar pa raha tho isko hum timeframe change karta ha
# daily sa weekly 
google_df['Volume'].resample(rule='ME').mean().plot()
```

```python
google_df['Volume'].resample(rule='QE').mean().plot()
```

```python
google_df['High'].plot()
```

```python
google_df['Volume'].resample(rule='YE').mean().plot()
```

```python

# humna ab monthly trend kar liya 
google_df['High'].resample('ME').mean().plot()
```

```python
google_df['High'].resample('YE').mean().plot()
```

### Upsampling
move from lower timeframe to upper timeframe

example apka data ka frequency `days` ma ha 
apna usko `month` ma convert kar diya using asfrequency  

```python
google_df['Close'].resample('h',).interpolate(method='spline',order=2).plot()
```

```python
# apna bola mera pass daily data ha hourly par jana ha
google_df['Close'].resample('h',).interpolate(method='linear').plot()
```

```python
# upsampling karo ga number of rows increase hogi aor nan values inc hogi fir aap unko fill karaga
```

```python
# 99 % cases ma aap downsample karta ho upsampling bhi kabhi kabhi ho sakta ha 
```

# Rolling Window(Smoothing)
Time series data in original format can be quite volatile, especially on smaller aggregation levels. The concept of rolling, or moving averages is a useful technique for smoothing time series data.

two types of moving average
- equal weighted to all -> moving avg
- exponetial moving weighted avg -> recent wala ko thoda jada weighted ,ussa picha wala ko thoda kam ,so on...

```python
# it is kind of similiar operation you study earlier
# moving average concept 
# days     some quantity
#  1          10
#  2          20
#  3          30
#  4          40
#  5          50
# ....
```

```python
# normal average aap total numbers /divide by it 
# lakin moving avg ma aap on the go avg nikalta ho
# example moving average = 3
# yani 3 days ka moving avg fir next ,fir next 
```

```python
# days     some quantity   moving avg of 3
#  1          10               nan
#  2          20               nan
#  3          30               avg(10,20,30)=20
#  4          40               avg(20,30,40)= 30
#  5          50               avg(30,40.50)
#                              previous 3 in all
# ....
```

```python
pd.Series([10,20,30,40,50]).rolling(3).mean()
```

```python

google_df['High'].rolling(50).mean().plot()
```

```python
# one year moving avg 
google_df['High'].rolling(365).mean().plot()
```

```python
google_df['High'].rolling(50).mean()
```

```python
# purana din yaad aa gaya trading ma buy or sell it 
google_df['High'].plot()

google_df['High'].rolling(50).mean().plot()
google_df['High'].rolling(200).mean().plot()
```

```python
# purana din yaad aa gaya trading ma buy or sell it 
google_df['High'].plot()

google_df['High'].rolling(365).mean().plot() # 1 year
google_df['High'].rolling(1095).mean().plot() # 3 year 
```

```python
# exponentail moving average 
google_df['High'].ewm(50).mean().plot()
```

```python
# exponentail moving average
google_df['High'].plot() 
google_df['High'].ewm(200).mean().plot()
```
