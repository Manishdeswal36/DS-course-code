### split-get

```python
# so now we have to make three coloumns 
# first name 
# surname
# title(Mr. Mrs.)

df['Name'].str.split(',')
# comma ka baad firstname and surname ha 
# hamana split kar diya uska name surname alg ho gaya
# promblem vo list ban gaya tho how we fetch the first item in it
df['Name'].str.split(',').str.get(0) # first name 
df['Name'].str.split(',').str.get(1) # sur name 
```

```python
df['lastname']= df['Name'].str.split(',').str.get(0)
```

```python
# always remember yadi apko sabhi rows par string operation lagan ha than use str accesssor 
```

```python
df['Name'].str.split(',').str.get(1).str.split(' ')
# promblem their is spaces in between 
```

```python
df['Name'].str.split(',').str.get(1).str.strip().str.split(' ')
# now kisika name Mr. Owen Harris ha tho split ho raha 
#[Mr., Owen, Harris]
# yani combined word hona tha lakin wo two words ma split ho gaya 
```

```python
# again split on the basis of space 
# very good use of brain split on the basis of dot 
df['Name'].str.split(',').str.get(1).str.strip().str.split('.')
# n yani kitni baar split karna 
# n= 1 phala space par split hoga
# n= 2 dursa space par split hoga 
df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1)
```

```python
df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)
# it created the dataframe by expand_true parameter 
```

```python
# now create a dataframe 
df[['title','firstname']]= df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)
```

```python
df.head()
```

```python
# kitna mr and kitna miss ha 
df['title'].value_counts()
```

### replace
ya replace karta ha ek string ko dursa string ma 

```python
# so Miss. and Ms. are same
# also Mile. jo ha french ma bola ha uskma matalab Miss. hi ha
df['title'] = df['title'].str.replace('Ms.',"Miss.")
df['title'] = df['title'].str.replace('Mlle.',"Miss.")
```

```python
df['title'].value_counts()
```

# filtering 

### startwith/endwith

```python
# aap filter kar sakta ho string ma
# returns --> boolean series 
df['firstname'].str.startswith('A') 
df[df['firstname'].str.startswith('A') ]
```

```python
df['firstname'].str.endswith('b')
df[df['firstname'].str.endswith('b')]
```

### isdigit/isalpha...

```python
# kiya string ma digit ha 
df['lastname'].str.isdigit()
df[df['lastname'].str.isdigit()]
```

```python
# ya ho gaya basic level filtering 
# aap advance level filtering bhi kar sakta ho .
# regex are complex pattern jis sa aap string ma searching kar sakta ho .
```

# apply regex

ya bhi filtering karna ha tarika ha string ma by pattern of string 

```python
# huma vo passenger chaya jinka name ma 'john' ata ha 
# john capital or small letter bhi chalaga
df['firstname'].str.contains('john')
df[df['firstname'].str.contains('john',case=False)]
# capital small dono chalega
```

### find last names with start and end char vowel

```python
# aapko vo person chaya jinka first name and last name vowel sa start or end hota ha
df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')
df[df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
```
