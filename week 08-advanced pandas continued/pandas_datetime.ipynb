3. tho weakness of python datetime format sa inspired ho kar numpy team  add a native time series data type to numoy

# date_range function

## generate daily dates in given_range

```python
pd.date_range(
    start='2025-1-5',
    end= '2025-1-31',
    freq='D'
    )
```

```python
# ek ek din chod kar 
pd.date_range(
    start='2025-1-5',
    end= '2025-2-28',
    freq='2D' # change the frequency 
    )
```

```python
# 3 days gaps 

pd.date_range(
    start='2025-1-5',
    end= '2025-2-28',
    freq='3D' # change the frequency 
    )
```

## business days 
yani sat and sun ko choti baki sara days 

```python
# pandas are intelligent 
# ki monday to friday par rakna ha 

pd.date_range(
    start='2025-1-5',
    end= '2025-2-28',
    freq='B' # change the frequency 
    )
```

## W-> one week per day

```python
# ya sara days jo ha vo sunday ha

pd.date_range(
    start='2025-1-5',
    end= '2025-2-28',
    freq='W' # change the frequency 
    )
```

```python
# apko sunday nahi thursday chaya
pd.date_range(
    start='2025-1-5',
    end= '2025-2-28',
    freq='W-THU' # change the frequency 
    )
```

## hourly -> timestamp 

```python
# apko din nahi har ghanta ka timestamp chaya
pd.date_range(
    start='2023/1/15',
    end='2023/2/28',
    freq='h'
)
```

```python
# apko din nahi har ghanta ka timestamp chaya
pd.date_range(
    start='2023/1/15',
    end='2023/2/28',
    freq='6h'
)
```

## M -> Month end

```python
pd.date_range(
    start='2023/1/15',
    end='2023/10/28',
    freq='ME'
)
```

## y -> Year start

```python
pd.date_range(
    start='2023/1/15',
    end='2025/2/28',
    freq='Y'
)
```

## YE -> year end 

```python

pd.date_range(
    start='2023/1/15',
    end='2025/2/28',
    freq='YE'
)
```

## periods 

```python
pd.date_range(
    start='2025/4/1',
    periods=25,# muja next 15 days dikoho
    freq='D'
)
```

```python
# next 25 hours diko 
pd.date_range(
    start='2025/4/1',
    periods=25,# muja next 15 days dikoho
    freq='h'
)
```

```python
# next 6 hours diko 
pd.date_range(
    start='2025/4/1',
    periods=25,# muja next 15 days dikoho
    freq='6h'
)
```

```python
# next month diko 
pd.date_range(
    start='2025/4/1',
    periods=25,# muja next 15 days dikoho
    freq='ME'
)
```

# to_datetime function
converts an existing objects to pandas timestamp/datetime64 objects

```python
a_string = pd.Series(['2025/1/1','2025/1/2','2025/1/3'])
a_string
# data type object ha
# inka aap fayda nahi udha sakta example inma sa year nikal kar dikho 
# apko split waraga karna padega 
```

```python
a_string.str.split("/").str.get(0)
# bhut jada mehnat 
```

## pd.to_datetime

```python

pd.to_datetime(a_string)
```

## dt accesor 

```python
# you also need a dt accessor
date_data = pd.to_datetime(a_string)
date_data.dt.year
```

```python
date_data.dt.month
```

```python
date_data.dt.day
```

```python
date_data.dt.month_name()
```

```python
date_data.dt.day_name()
```

## with errors 

```python
a_string = pd.Series(['2025/1/1','2025/1/2','2025/130/3'])

# man lo kisi na 130 month ma dal diya
pd.to_datetime(a_string)
# ab vo crash erro dega 
```

```python
# to solve this error use coerce yani skip kar do un values ko 
pd.to_datetime(a_string,errors='coerce')
# ab yaha aa gaya NaT yani not a date time 
```

```python
# now you use all operations you use before
pd.to_datetime(a_string,errors='coerce').dt.month_name()
```

# example 

```python
expense_df = pd.read_csv('expense_data.csv')
expense_df.head()
```

```python
expense_df.info()
```

```python
# app isma datetime object ko use kar sakta until aap isko datatime ma convert na karo 
expense_df['Date']= pd.to_datetime(expense_df['Date'])
```

```python
expense_df.info()
```

## dt.accessor 

```python
expense_df['Date'].dt.month_name()
```

```python
expense_df['Date'].dt.is_month_end
```

```python
# kiya ya quarter end ha 
# and many types of operation like this 
expense_df['Date'].dt.is_quarter_end
```

## plot graph 

```python
import matplotlib.pyplot as plt
plt.plot(expense_df['Date'],expense_df['INR'])
```

```python
# ya ha har date ko kitna karcha kiya 
```

```python
# ab apko day name wise chart/month wise bar chart
# sabsa jada karcha konsa din karta ha monday,tues,wed kab
expense_df['Date'].dt.day_name()
```

```python
expense_df['day_name']= expense_df['Date'].dt.day_name()
```

```python
expense_df.head()
```

```python
expense_df.groupby('day_name')['INR'].sum().sort_values(ascending=False).plot(kind='bar')
```

```python
expense_df['month_name'] = expense_df['Date'].dt.month_name()
```

```python
expense_df.head()
```

```python
expense_df.groupby('month_name')['INR'].sum().plot(kind='bar')
```

```python
# month ka end ma karcha 
expense_df[expense_df['Date'].dt.is_month_end][['Date','INR']].sort_values(by='Date').set_index(keys='Date').plot(kind='bar')
```
